# FR06 — İnvestisiyalar və fiskal ötürücü blok

İnvestisiyaların strukturu və real artım sürətlərinin proqnozlaşdırılması (MİİS §15.5.1, bənd
2.4.1.5.-7.), üstəgəl büdcə kapital xərclərinin investisiyaya ötürülmə kanalı (fiskal blok və investisiya deflyatoru).

## 1. Tələb

**Spesifikasiyanın mətni (§15.5.1, FR6):** "İnvestisiyaların strukturu və real artım sürətlərinin
proqnozlaşdırılması."

Tələb iki qatlıdır: (a) investisiyanın **strukturu** — daxili/xarici, dövlət/qeyri-dövlət və
neft-qaz/qeyri-neft-qaz bölgüləri üzrə paylar, və (b) hər bir komponentin **real artım tempi**.
Paketin memarlıq qaydasına görə (BUILD_CONTEXT §D2, §D3) bu dəftər icra DAG-ının dördüncü
mərhələsindədir və FR01-in artıq həll etdiyi simultan nüvədən faydalanır.

Nüvənin investisiya tənliyi nəşr olunmuş **cəmi** investisiya artımı sırası (`inv_total_g`)
üzərində qiymətləndirilib və FR01-in tələb tənliyinə də məhz cəmi investisiya artımı kimi daxil
olur. Buradan bu dəftərin qapanma qaydası çıxır: **təhvil verilən cəmi investisiya yolu FR01
nüvəsinin yolunun ÖZÜdür** — `g_inv_core` 2025-ci ilin cəmi investisiya səviyyəsinə tətbiq olunur,
ayrıca qiymətləndirilmir və heç bir düzəlişə məruz qalmır. Xarici investisiya (FDI) ekzogen
fərziyyədir, **daxili investisiya isə qalıq kimi** alınır: daxili = cəmi − xarici.

**Bu buraxılışda dəyişən dörd şey (fiskal blok və investisiya deflyatoru).**

| Element | Əvvəlki buraxılış | Bu buraxılış |
|---|---|---|
| Dövlət/qeyri-dövlət bölgüsü | son üç ilin sabit payı (56,47 %), cəmi yola vurulur | **büdcənin əsaslı vəsait qoyuluşu (DƏVQ) tənliyə daxil olur** — dövlət investisiyası tənliyinin iki termli konstruksiyası qurulur, sınağa buraxılır və çəkisi sınaqla təyin olunur |
| Neft/qeyri-neft bölgüsü | sabit pay (27,07 %) | neft-qaz blokunun fəaliyyət göstəriciləri ilə **modelləşdirilir**; sabit pay yarışda saxlanılır və nəticəsi nəşr olunur |
| Qeyri-neft × dövlət/qeyri-dövlət | **yoxdur** | `real_r102` (2018–2025) ilə qurulur; dövlət hissəsi eynilikdir |
| İnvestisiya deflyatoru | **yoxdur** — hər şey 2025 sabit qiymətləri ilə | implisit investisiya deflyatoru qurulur, Nazirliyin `eq_defid` tənliyi bizim qatda yenidən qiymətləndirilir, qazanan yol nominal körpüyə bağlanır |

Əlavə olaraq dəftər **fiskal ötürücü blokun** iki nəticəsini nəşr edir: büdcənin əsaslı vəsait
qoyuluşunun (DƏVQ) sırası və dövlət büdcəsi üzrə **qeyri-neft balansı göstəricisi**. İkincisi
icmal büdcə perimetri DEYİL və etiketində bu açıq yazılır (fiskal kanal 3).

## 3. Model seçimi

### 3.1 Cəmi investisiya — FR01 nüvəsinin yolu, düzəlişsiz

FR01-in simultan nüvəsi (§D3) dörd tənlikdən ibarətdir; onlardan biri — kredit kanalı vasitəsilə
investisiyanın real artımını izah edən tənlik — bilavasitə bu dəftərin mövzusudur. Tənlik nəşr
olunmuş **cəmi** investisiya artımı sırası (`inv_total_g`, 2001–2025) üzərində qiymətləndirilib və
FR01-in tələb tənliyinə də cəmi investisiya artımı kimi daxil olur. Ona görə burada həmin yol
**cəmi** investisiyaya tətbiq edilir. Bu, seçim deyil, uyğunluq tələbidir.

### 3.2 Xarici investisiya — ekzogen, Nazirliyin REAL artım yolu ilə

Geriyə doğru sınaq AR(1), beş-illik orta və neft qiymətinə əsaslanan modelləri RW etalonuna qarşı
sınaqdan keçirir; ən yaxşı sağ qalan ansambl RW-ni cəmi ~11 % üstələyir, bu isə ~27–30 f.b. mütləq
RMSE-yə nisbətdə praktik əhəmiyyətsizdir. Xarici investisiya Azərbaycanda iri, kəsilməz olmayan
neft-qaz layihələrinin cədvəli ilə idarə olunur — bir dəyişənli zaman sırası modelinin tuta
bilmədiyi məlumatdır. Ona görə mərkəzi yol Nazirliyin öz FDI layihə baza ssenarisi əsasında
**ekzogen fərziyyə** kimi qəbul edilir.

**Vintaj düzəlişi.** Şablon iki sətir verir: nominal **səviyyə** (sətir 16) və **real artım tempi**
(sətir 17). Səviyyə sütunu Nazirliyin öz 2025 qiymətləndirməsi üzərində qurulub: 2026 üçün
6 080,1 mln AZN həmin ilin +16,4 % real artımı ilə 5 222,6 mln AZN-lik implisit 2025 bazası verir,
iş kitabının nəşr olunmuş 2025 dəyəri isə 4 203,0 mln AZN-dir — **vintaj fərqi 24,3 %**. Nominal
səviyyəni faktiki dəyərin üzərinə zəncirləmək 2026 üçün +44,7 %-lik saxta "real artım" yaradardı.
Ona görə **real artım tempi** sətri götürülür və iş kitabının öz 2025 səviyyəsinə tətbiq olunur.

### 3.3 Daxili investisiya — qalıq

Daxili = cəmi − xarici (hər ikisi 2025 sabit qiymətləri ilə). Tənlik cəmi sıra üzərində
qiymətləndirilib, ona görə cəmi sıraya tətbiq olunur; FDI-ın öz layihə cədvəli var, ona görə
ekzogendir; qalan hissə isə qalıqdır.

### 3.4 Dövlət/qeyri-dövlət — dövlət investisiyası tənliyinin iki termli konstruksiyası

Əvvəlki buraxılışda pay sabit idi (son üç ilin ortası). İcmal büdcə kapital xərclərinin
tənliyə daxil edilməsini tələb edir. Bölmə 2.2-nin sübutu göstərir ki, **tək başına** DƏVQ artıq
kifayət deyil: onun dövlət investisiyasındakı payı 2018-ci ildəki 60 %-dən 2025-ci ildə 20,4 %-ə
enib. Ona görə iki termli forma qurulur:

```
Δln I^dövlət = α + β₁·Δln(DƏVQ, real) + β₂·Δln(SOE öz vəsaitləri, real) + ε
```

**İkinci term BU BURAXILIŞDA qiymətləndirilir (bu buraxılış).** Əvvəlki buraxılışda `real_r114`-ün
illik sütunu yalnız 2023–2025 üçün dolu idi (n = 3) və D5-in nümunə intizamına görə həmin sıra
üzərində elastiklik nəşr edilmirdi. Açıq mənbə sweep-i eyni anlayışın **1999–2024 tarixçəsini**
tapdı (opendata.az, maliyyə mənbələri üzrə investisiyalar); iki sıra üst-üstə düşən iki ildə
**dəqiq eynidir**, ona görə birləşdirilmiş sıra 1999–2025 dövrünü əhatə edir və term artıq
qiymətləndirilə biləndir. Beləliklə:

* **birinci term** (DƏVQ) tam qiymətləndirilir və **qapıdan** keçirilir (p < 0,10 və düzgün işarə);
* **ikinci term** iki formada verilir: (a) əsl `öz vəsaitləri` sırası ilə — iki termli formanın nəzərdə
  tutduğu forma, artıq **nəşr olunan əmsalla**; (b) əvvəlki buraxılışın sənədləşdirilmiş
  ƏVƏZEDİCİsi ilə — neft-qaz sektoruna investisiya (`real_r082`), müqayisə üçün saxlanılır;
* hər iki forma eyni pəncərədə, eyni etalona qarşı **yarışa** buraxılır və nəticə — udmasından
  asılı olmayaraq — həm tənlik kataloqunda, həm sınaq cədvəlində qalır.

**Nümunəni məhdudlaşdıran indi öz vəsait sırası deyil, DƏVQ-dir:** büdcənin əsaslı vəsait qoyuluşu
sırası 2010-cu ildən başlayır, ona görə iki termli tənliyin ortaq nümunəsi n = 15-dir. Öz vəsait
termi TƏK BAŞINA isə tam 26 illik nümunədə qiymətləndirilir və nəticəsi ayrıca göstərilir.

Qərar yenə sınağındır: dövlət investisiyası tənliyi, v2-nin üç illik pay qaydası, AR(1), beş-illik orta və təsadüfi
gəzişmə eyni pəncərədə yarışır; mərkəzi yolu ansambl daşıyır və **hər modelin çəkisi sınaqdan
gəlir**, iqtisadi mülahizədən yox.

### 3.5 Neft/qeyri-neft — sabit pay yarışa buraxılır

Əvvəlki buraxılışda pay sabit idi (27,07 %). Bu buraxılışda neft-qaz sektoruna investisiyanın öz
real artımı **neft-qaz blokunun fəaliyyət göstəriciləri** ilə izah edilməyə çalışılır: çəkili
neft+qaz ixrac qiyməti indeksi (`oilgas_pw_g`), əmtəəlik həcm indeksi (`oil_realg`) və Brent.
Gözlənilən işarə **müsbətdir** (qiymət/fəaliyyət artdıqca investisiya artır) — bu, qapının bir
hissəsidir və nəticə nə olursa olsun dəyişdirilmir. Sabit pay qaydası da eyni yarışa girir.

### 3.6 İnvestisiya deflyatoru (investisiyanın ayrıca qiymət indeksi)

İcmal investisiyanın ayrıca qiymət indeksinin olmamasını qeyd edir. Bu buraxılışda:

* **hədəf sıra bizim öz qatımızda qurulur** — `real_r080` (nominal səviyyə) və `real_r081`
  (əvvəlki il = 100 real indeks) qonşu sətirlərindən implisit deflyator. 2016-cı il dəyəri
  **26,24 %** MOE SNA `Investisiya` vərəqinin öz "deflyator" sətri ilə (26,23964) beş onluq rəqəmə
  qədər üst-üstə düşür — yəni boşluq bizim qatda bağlanır;
* **Nazirliyin `eq_defid` tənliyi olduğu kimi qəbul edilir və yenidən qiymətləndirilir**:
  öz gecikməsi + gecikmiş NEER + rejim sabitləri;
* **tikinti deflyatoru ilə idxal qiymətlərinin çəkili düsturu** — `w_c·(tikinti-quraşdırma deflyatoru) + (1−w_c)·(idxal qiymətləri)`,
  `w_c` = tikinti işlərinin cəmi investisiyadakı payı — həm **tətbiq olunmuş çəki** ilə, həm də
  **qiymətləndirilmiş çəki** ilə yarışa daxil edilir;
* qazanan yol nominal körpüyə bağlanır: `inv_total_nom` üçün proqnoz sətirləri ilk dəfə nəşr olunur
  və `price_basis` sütunu hər sırada öz bazasını elan edir.

### 5.1 Cəmi investisiya — kredit kanalı

FR01-in konstruksiyasını təkrarlayaraq real M2 kütləsinin İQİ-yə görə deflyasiya edilmiş artımı
(`dlrm2` = Δln(M2/İQİ) × 100) qurulur; 2015–2016 devalvasiyası üçün süni dəyişən (`d15`) əlavə olunur.

````python
cpi_ix = pd.Series(dtype=float); cpi_ix.loc[2015] = 1.0
for y in range(2016, LA + 1):
    cpi_ix.loc[y] = cpi_ix.loc[y - 1] * (1 + cpi.loc[y] / 100.0)
for y in range(2014, int(cpi.index.min()) - 1, -1):
    if (y + 1) in cpi.index:
        cpi_ix.loc[y] = cpi_ix.loc[y + 1] / (1 + cpi.loc[y + 1] / 100.0)
cpi_ix = cpi_ix.sort_index()

dlrm2 = (np.log(m2 / cpi_ix.reindex(m2.index)).diff() * 100).rename("dlrm2")

d15 = pd.Series(0.0, index=range(1995, config.LAST_FORECAST + 1), name="d15")
d15.loc[[2015, 2016]] = 1.0

S = pd.DataFrame({"y": inv_total_g, "dlrm2": dlrm2, "dlrm2_l1": dlrm2.shift(1), "d15": d15}).loc[2000:LA]
S.round(2)
````

````python
sI = S.loc[2008:LA]
res_inv = ek.ols_hac(sI["y"], sI[["dlrm2", "dlrm2_l1", "d15"]], maxlags=1,
                     name="STRUKTUR: kredit kanalı, gecikməli, süni dəyişənli (istifadə olunur)")

sIu = S.loc[2007:LA].dropna(subset=["y", "dlrm2"])
res_inv_nd = ek.ols_hac(sIu["y"], sIu[["dlrm2"]], maxlags=1,
                        name="ROBUSTLİK: süni dəyişənsiz, gecikməsiz")
res_inv_d = ek.ols_hac(sIu["y"], sIu[["dlrm2", "d15"]], maxlags=1,
                       name="ROBUSTLİK: süni dəyişənli, gecikməsiz")

sIx = sIu.drop(index=[y for y in (2015, 2016) if y in sIu.index])
res_inv_x = ek.ols_hac(sIx["y"], sIx[["dlrm2"]], maxlags=1,
                       name="ROBUSTLİK: 2015-2016 nümunədən çıxarılıb, süni dəyişənsiz")

print(res_inv); print()
print(res_inv_nd); print()
print(res_inv_d); print()
print(res_inv_x)
````

### 5.2 2015 hekayəsi

Dörd spesifikasiya bir yerdə oxunanda 2015–2016 devalvasiyasının kredit kanalına təsiri aydın
görünür. 2015-ci ildə real M2 kütləsi 74 f.b.-ə yaxın kəskin **daralıb** (`dlrm2` = −73,7), halbuki
investisiya artımı bundan qat-qat az kəskin düşüb (−11,1 %). Gecikməli, süni dəyişənli spesifikasiya
bu uyğunsuzluğu düzgün idarə edən yeganə formadır: `dlrm2` əmsalı 0,552, gecikmiş həddi 0,646, süni
dəyişən +30,61 — hamısı 1 % səviyyəsində əhəmiyyətlidir. Gecikmə həddi çıxarıldıqda əmsal 0,292-yə,
düzəldilmiş R² 0,167-yə enir; süni dəyişən gecikməsiz daxil edildikdə isə işarəsi tərsinə dönür.
2015–2016 nümunədən tamamilə çıxarıldıqda əmsal 0,563-ə qayıdır — yəni devalvasiya illərini ya
düzgün gecikmə strukturu ilə, ya da nümunədən çıxarmaqla idarə etmək olar; onları naiv şəkildə
daxil etmək əmsalı qeyri-sabit edir.

````python
from IPython.display import Math, display


def _eq_latex(res, labels, lhs):
    body = ""
    for i, nm in enumerate(res.exog_names):
        b, se, lab = res.params[nm], res.se[nm], labels[nm]
        pre = ("-" if b < 0 else "") if i == 0 else (" - " if b < 0 else " + ")
        term = f"{abs(b):.3f}" + ("" if lab == "" else r"\," + lab)
        body += pre + r"\underset{(" + f"{se:.3f}" + r")}{" + term + "}"
    return lhs + " = " + body


LBL = {"const": "", "dlrm2": r"\Delta\ln(M2/P)_t", "dlrm2_l1": r"\Delta\ln(M2/P)_{t-1}",
       "d15": r"D^{15\text{-}16}_t"}
display(Math(_eq_latex(res_inv, LBL, r"\hat{g}^{I}_t")))
print(f"n = {res_inv.n},  nümunə = {res_inv.sample[0]}–{res_inv.sample[1]},  "
      f"düzəldilmiş R² = {res_inv.adj_r2:.3f},  reqressiyanın standart xətası = {res_inv.sigma:.3f}")
print("mötərizədə HAC (Newey–West, gecikmə 1) standart xətaları verilib.")
````

### 5.3 Dövlət investisiyası — iki termli tənlik

`DƏVQ` və neft-qaz investisiyası İQİ ilə deflyasiya olunur (hər ikisi nominal sıradır), asılı
dəyişən isə nəşr olunmuş **real** artım tempidir (`inv_state_g`). Aşağıda dörd spesifikasiya
verilir: tək termli, tək termli + devalvasiya süni dəyişəni, iki termli (əvəzedici ilə) və
**iki termli, `real_r114` ilə** — sonuncusu nümunə ölçüsünü GÖSTƏRMƏK üçündür, nəticəsi
nəşr edilmir.

````python
def real_g(s):
    """nominal səviyyə sırasını İQİ ilə deflyasiya edib real artım tempinə (%) çevirir"""
    s = pd.Series(s).astype(float).dropna().sort_index()
    g = (s / s.shift(1) - 1) * 100
    return (((1 + g / 100) / (1 + cpi.reindex(g.index) / 100) - 1) * 100).dropna()


def _cut(s):
    s = pd.Series(s).astype(float).sort_index()
    return s[s.index <= LA]


g_capex = real_g(capex).rename("g_capex")                       # dövlət investisiyası tənliyinin birinci termi
g_oilinv = real_g(_cut(data_layer.wb_labelled("real_r082"))).rename("g_oilinv")
g_soe = real_g(soe).rename("g_soe")                             # dövlət investisiyası tənliyinin ikinci termi — TAM tarixçə (1999-2025)

ST = pd.DataFrame({"y": inv_state_g, "devq_realg": g_capex, "neftinv_realg": g_oilinv, "oz_vesait_realg": g_soe,
                   "d15": d15, "gtot": inv_total_g}).loc[:LA]

_s1 = ST[["y", "devq_realg"]].dropna()
res_st1 = ek.ols_hac(_s1["y"], _s1[["devq_realg"]], maxlags=1,
                     name="DİT-1: dövlət investisiyası ~ DƏVQ (real)")
_s2 = ST[["y", "devq_realg", "d15"]].dropna()
res_st2 = ek.ols_hac(_s2["y"], _s2[["devq_realg", "d15"]], maxlags=1,
                     name="DİT-2 (istifadə olunur): DƏVQ + 2015-16 süni dəyişəni")
_s3 = ST[["y", "devq_realg", "neftinv_realg"]].dropna()
res_st3 = ek.ols_hac(_s3["y"], _s3[["devq_realg", "neftinv_realg"]], maxlags=1,
                     name="DİT-3: iki termli — DƏVQ + neft-qaz investisiyası (əvəzedici)")
print(res_st1); print()
print(res_st2); print()
print(res_st3); print()
_s4 = ST[["y", "devq_realg", "oz_vesait_realg"]].dropna()
res_st4 = ek.ols_hac(_s4["y"], _s4[["devq_realg", "oz_vesait_realg"]], maxlags=1,
                     name="DİT-4: iki termli — DƏVQ + ÖZ VƏSAİTLƏRİ (iki termli formanın əsl variantı)")
print(res_st4); print()
_s5 = ST[["y", "oz_vesait_realg"]].dropna()
res_st5 = ek.ols_hac(_s5["y"], _s5[["oz_vesait_realg"]], maxlags=1,
                     name="DİT-5: yalnız öz vəsaitləri — TAM nümunə (DƏVQ ilə məhdudlaşmır)")
print(res_st5)
print()
print(f"Nümunə ölçüləri: iki termli n = {len(_s4)} ({int(_s4.index.min())}-{int(_s4.index.max())}, "
      f"DƏVQ sırası ilə məhdudlaşır); yalnız öz vəsaitləri n = {len(_s5)} "
      f"({int(_s5.index.min())}-{int(_s5.index.max())}).")
````

````python
# QAPI (sahə sürücüsü qapısı ilə eyni): p < 0,10 VƏ gözlənilən işarə (DƏVQ üçün müsbət).
GATE_P = 0.10
b_st, p_st = float(res_st2.params["devq_realg"]), float(res_st2.pvalues["devq_realg"])
gate_state = bool(p_st < GATE_P and b_st > 0)
print(f"QAPI — dövlət investisiyası ~ DƏVQ: β = {b_st:+.3f}, p = {p_st:.4f}, "
      f"gözlənilən işarə müsbət → {'KEÇİR' if gate_state else 'KEÇMİR'}")
b_st3, p_st3 = float(res_st3.params["neftinv_realg"]), float(res_st3.pvalues["neftinv_realg"])
print(f"QAPI — ikinci term (neft-qaz investisiyası əvəzedicisi): β₂ = {b_st3:+.3f}, "
      f"p = {p_st3:.4f} → {'KEÇİR' if (p_st3 < GATE_P and b_st3 > 0) else 'KEÇMİR'}")
b_st4, p_st4 = float(res_st4.params["oz_vesait_realg"]), float(res_st4.pvalues["oz_vesait_realg"])
gate_soe = bool(p_st4 < GATE_P and b_st4 > 0)
print(f"QAPI — ikinci term (ƏSL öz vəsait sırası): β₂ = {b_st4:+.3f}, p = {p_st4:.4f} → "
      f"{'KEÇİR' if gate_soe else 'KEÇMİR'}")
b_st5, p_st5 = float(res_st5.params["oz_vesait_realg"]), float(res_st5.pvalues["oz_vesait_realg"])
print(f"QAPI — yalnız öz vəsaitləri (n={res_st5.n}): β = {b_st5:+.3f}, p = {p_st5:.4f} → "
      f"{'KEÇİR' if (p_st5 < GATE_P and b_st5 > 0) else 'KEÇMİR'}")
````

**dövlət investisiyası tənliyinin ikinci termi haqqında nəticə — açıq və mənfi.** Sıra tapıldı, tənlik quruldu, əmsal qiymətləndirildi: iki termli formada öz vəsaitlərin real artımının əmsalı **sıfırdan ayırd edilə bilmir** (p ≈ 0,95), düzəldilmiş R² isə tək termli formanın 0,263-dən 0,202-yə **enir** — yəni term modelə məlumat gətirmir, sərbəstlik dərəcəsi aparır. Tam 26 illik nümunədə tək başına qiymətləndirildikdə də əlaqə yoxdur (β = −0,104, p = 0,73).

Bu, məlumat çatışmazlığı deyil, **ölçülmüş nəticədir** və səbəbi §2.4-də artıq qeyd olunmuş etiket fərqidir: sıra *maliyyələşmə mənbəyi* üzrə bölgüdür (bütün müəssisələrin öz vəsaitləri), *mülkiyyət* üzrə deyil — onun böyük hissəsi qeyri-dövlət müəssisələrinin öz vəsaitidir və dövlət investisiyası ilə eyni istiqamətdə hərəkət etməsi üçün heç bir səbəb yoxdur. Term buna baxmayaraq §6.3-ün yarışına **daxil edilir**: paketin qaydası tənliyin taleyini t-statistikası deyil, sınaq həll edir (D4).

## 6. Geriyə doğru sınaqlar

Beş ayrı yarış aparılır, hamısı genişlənən pəncərəli, bir addımlıq, nümunədən kənar prosedura görə
(§D4): (a) kredit kanalı — cəmi investisiya; (b) xarici investisiya; (c) **dövlət investisiyası**
— dövlət investisiyası tənliyi v2-nin üç illik pay qaydasına və RW-yə qarşı; (d) **neft-qaz investisiyası** — neft
blokunun sürücüləri sabit paya və RW-yə qarşı; (e) **investisiya deflyatoru** — Nazirliyin öz
tənliyi, çəkili proxy düsturu və bizim qatın spesifikasiyası RW-yə qarşı. Uduzan model silinmir:
hər birinin RMSE-si `validation_backtest.csv`-ə yazılır.

````python
import statsmodels.api as sm


def _fit(train, cols, need=9):
    d = train[["y"] + cols].dropna()
    if len(d) < need:
        return None
    return sm.OLS(d["y"], sm.add_constant(d[cols], has_constant="add")).fit()


def m_struct(train, t):
    f = _fit(train, ["dlrm2", "dlrm2_l1", "d15"])
    if f is None or t not in S.index or pd.isna(S.loc[t, "dlrm2"]) or pd.isna(S.loc[t, "dlrm2_l1"]):
        return np.nan
    r = S.loc[t]
    return float(f.params["const"] + f.params["dlrm2"] * r["dlrm2"]
                 + f.params["dlrm2_l1"] * r["dlrm2_l1"] + f.params["d15"] * r["d15"])


def m_simple(train, t):
    f = _fit(train, ["dlrm2"], need=8)
    if f is None or t not in S.index or pd.isna(S.loc[t, "dlrm2"]):
        return np.nan
    return float(f.params["const"] + f.params["dlrm2"] * S.loc[t, "dlrm2"])


def m_ar1(train, t):
    d = pd.DataFrame({"y": train["y"], "l1": train["y"].shift(1)}).dropna()
    if len(d) < 6:
        return np.nan
    f = sm.OLS(d["y"], sm.add_constant(d[["l1"]], has_constant="add")).fit()
    return float(f.params["const"] + f.params["l1"] * train["y"].dropna().iloc[-1])


def m_mean5(train, t):
    s = train["y"].dropna()
    return float(s.iloc[-5:].mean()) if len(s) >= 5 else np.nan


bt = ek.backtest(S, {"STRUKTUR": m_struct, "STRUKTUR_SADE": m_simple, "AR1": m_ar1, "ORTA5": m_mean5},
                 min_train=10, series_code="inv_total_g", verbose=True)
bt.table.groupby("model")["rmse_h"].first().rename("RMSE")
````

**Xarici investisiya üçün namizədlər:** AR(1), son beş ilin ortası, neft qiymətinə əsaslanan xətti
reqressiya (`brent_usd`, cari il).

````python
frame_for = pd.DataFrame({"y": inv_for_g, "brent": brent.reindex(inv_for_g.index)})


def m_brent(train, t):
    d = pd.DataFrame({"y": train["y"], "b": train["brent"]}).dropna()
    if len(d) < 8:
        return np.nan
    f = sm.OLS(d["y"], sm.add_constant(d[["b"]], has_constant="add")).fit()
    if t not in frame_for.index or pd.isna(frame_for.loc[t, "brent"]):
        return np.nan
    return float(f.params["const"] + f.params["b"] * frame_for.loc[t, "brent"])


bt2 = ek.backtest(frame_for, {"AR1": m_ar1, "ORTA5": m_mean5, "BRENT": m_brent},
                  min_train=8, series_code="inv_foreign_g", verbose=True)
bt2.table.groupby("model")["rmse_h"].first().rename("RMSE")
````

### 6.3 Dövlət investisiyası — dövlət investisiyası tənliyi v2-nin pay qaydasına qarşı

`V2_PAY3` v2-nin qaydasının DƏQİQ təkrarıdır: pay son üç ilin sadə ortasıdır, səviyyə isə cəmi
investisiyanın həmin ildəki dəyərinə vurulur. Hər iki model cari ilin bir sürücüsünü bilir (biri
DƏVQ-i, digəri cəmi investisiyanı), ona görə müqayisə şərtləri eynidir.

````python
def _mk_lin(frame, cols, need=9):
    def f(train, t):
        d = train[["y"] + cols].dropna()
        if len(d) < need:
            return np.nan
        fit = sm.OLS(d["y"], sm.add_constant(d[cols], has_constant="add")).fit()
        if t not in frame.index or frame.loc[t, cols].isna().any():
            return np.nan
        return float(fit.params["const"] + sum(fit.params[c] * frame.loc[t, c] for c in cols))
    return f


def _mk_share3(level_series, total_series):
    """v2 qaydası: pay = son üç ilin ortası, səviyyə = cəmi × pay, artım = səviyyə/keçən il − 1"""
    sh_all = (level_series / total_series * 100).dropna()

    def f(train, t):
        sh = sh_all[sh_all.index < t]
        if len(sh) < 3 or (t - 1) not in level_series.index or t not in total_series.index:
            return np.nan
        lvl = float(total_series.loc[t]) * float(sh.iloc[-3:].mean()) / 100.0
        return (lvl / float(level_series.loc[t - 1]) - 1) * 100
    return f


bt_state = ek.backtest(
    ST, {"DEVQ": _mk_lin(ST, ["devq_realg"]), "DEVQ_D15": _mk_lin(ST, ["devq_realg", "d15"]),
         "IKITERMLI": _mk_lin(ST, ["devq_realg", "neftinv_realg"]),
         # bu buraxılış: iki termli formanın əsl variantı — artıq qiymətləndirilə bilir
         "IKITERMLI_OZ_VESAIT": _mk_lin(ST, ["devq_realg", "oz_vesait_realg"]),
         "V2_PAY3": _mk_share3(inv_state, inv_total), "AR1": m_ar1, "ORTA5": m_mean5},
    min_train=9, series_code="inv_state_g", verbose=True)
display(pd.Series(bt_state.rmse, name="RMSE").round(3).to_frame()
        .assign(**{"RW-ni üstələyir": pd.Series(bt_state.beats_rw)}))
print(f"RW RMSE = {bt_state.rw_rmse:.3f}")
````

### Dinamiklik göstəricisi və dinamika formaları

**Dinamiklik göstəricisi** — σ(proqnoz 2026–2030)/σ(faktiki 2013–2025), hər ikisi populyasiya
standart kənarlaşmasıdır. `validation_backtest.csv`-ə `model = "DINAMIKLIK"` sətirləri ilə yazılır.
Şərti orta proqnoz tərifinə görə faktikidən az dəyişkəndir, ona görə 1-dən kiçik dəyər qüsur
deyil; sıfıra yaxın dəyər isə proqnozun praktik olaraq sabit olduğunu bildirir. Aralıqdan kənara
düşən HƏR sıra üçün yazılı əsaslandırma tələb olunur (D2) və əsaslandırma

### 7.2 Dövlət/qeyri-dövlət bölgüsü — ansambldan

Hər saxlanılan modelin öz yolu qurulur, sonra Bölmə 6.3-ün çəkiləri ilə birləşdirilir. Qeyri-dövlət
komponenti qalıq kimi alınır (cəmi − dövlət), ona görə eynilik hər il dəqiq qapanır.

````python
def _share3(level_series, total_series):
    sh = (level_series / total_series * 100).dropna()
    return float(sh.iloc[-3:].mean())


def _ar1_path(hist, years):
    d = pd.DataFrame({"y": hist, "l1": hist.shift(1)}).dropna()
    f = sm.OLS(d["y"], sm.add_constant(d[["l1"]], has_constant="add")).fit()
    out, last = {}, float(hist.dropna().iloc[-1])
    for t in years:
        last = float(f.params["const"] + f.params["l1"] * last)
        out[int(t)] = last
    return out


def _lin_path(frame, cols, drivers, years):
    d = frame[["y"] + cols].dropna()
    f = sm.OLS(d["y"], sm.add_constant(d[cols], has_constant="add")).fit()
    return {int(t): float(f.params["const"] + sum(f.params[c] * drivers[c][t] for c in cols))
            for t in years}


def _share_path(level_series, total_series, total_path, years):
    """v2 qaydası proqnoz üfüqündə: sabit pay × cəmi yolu, sonra artıma çevrilir"""
    share = _share3(level_series, total_series)
    prev = float(level_series.loc[LA])
    out = {}
    for t in years:
        cur = float(total_path.loc[t]) * share / 100.0
        out[int(t)] = (cur / prev - 1) * 100
        prev = cur
    return out


def _combine(bt_res, paths, years):
    """Ansambl çəkiləri ilə birləşdirir; heç bir model saxlanılmayıbsa RW yolunu qaytarır."""
    w = {m: v for m, v in bt_res.weights.items() if m in paths}
    if not w:
        return pd.Series({int(t): float(paths["RW"][int(t)]) for t in years})
    tot = sum(w.values())
    w = {m: v / tot for m, v in w.items()}
    return pd.Series({int(t): sum(w[m] * float(paths[m][int(t)]) for m in w) for t in years})


# İkinci termin (neft-qaz investisiyasının real artımı) proqnoz yolu üçün öz tarixçəsinin profil
# sınağı işlədilir — bu term mərkəzi yolu daşımır (yarışda uduzur), lakin ansamblın bir üzvü
# olduğu üçün yolu YENƏ DƏ göstərilməlidir; sabit dəyər qoymaq gizli fərziyyə olardı.
_x2_prof, _x2_bt, _x2_note = ek.profile_path(g_oilinv, FY, name="inv_oil_realg_profil",
                                             min_train=8, min_points=4)
# Öz vəsait termi üçün də üfüq yolu öz profil sınağından gəlir (sabit dəyər gizli fərziyyə olardı).
_x3_prof, _x3_bt, _x3_note = ek.profile_path(g_soe, FY, name="inv_own_funds_realg_profil",
                                             min_train=8, min_points=4)
print("öz vəsait investisiyasının real artımı —", _x3_note)
_drv_state = {"devq_realg": {int(y): float(capex_realg.loc[y]) for y in FY},
              "d15": {int(y): 0.0 for y in FY},
              "neftinv_realg": {int(y): float(_x2_prof.loc[y]) for y in FY},
              "oz_vesait_realg": {int(y): float(_x3_prof.loc[y]) for y in FY}}
state_paths = {
    "DEVQ": _lin_path(ST, ["devq_realg"], _drv_state, FY),
    "DEVQ_D15": _lin_path(ST, ["devq_realg", "d15"], _drv_state, FY),
    "IKITERMLI": _lin_path(ST, ["devq_realg", "neftinv_realg"], _drv_state, FY),
    "IKITERMLI_OZ_VESAIT": _lin_path(ST, ["devq_realg", "oz_vesait_realg"], _drv_state, FY),
    "V2_PAY3": _share_path(inv_state, inv_total, total_level, FY),
    "AR1": _ar1_path(inv_state_g, FY),
    "ORTA5": {int(y): float(inv_state_g.dropna().iloc[-5:].mean()) for y in FY},
    "RW": {int(y): float(inv_state_g.dropna().iloc[-1]) for y in FY},
}
state_g = _combine(bt_state, state_paths, FY).rename("inv_state_g")

state_level, prev = {}, float(inv_state.loc[LA])
for y in FY:
    prev = prev * (1 + float(state_g.loc[y]) / 100.0)
    state_level[y] = prev
state_level = pd.Series(state_level, name="inv_state")
nonstate_level = (total_level - state_level).rename("inv_nonstate")
nonstate_g = (nonstate_level / pd.concat([inv_nonst.loc[[LA]], nonstate_level]).shift(1).loc[FY] - 1) * 100
nonstate_g.name = "inv_nonstate_g"

print("dövlət investisiyası — ansambl çəkiləri:", {k: round(v, 3) for k, v in bt_state.weights.items()})
display(pd.DataFrame({m: pd.Series(p) for m, p in state_paths.items()}).round(2))
_sh_now = float(inv_state.loc[LA]) / float(inv_total.loc[LA]) * 100
_sh_path = (state_level / total_level * 100)
print(f"NƏTİCƏNİN STRUKTUR TƏSİRİ (gizlədilmir): dövlətin cəmi investisiyadakı payı "
      f"{_sh_now:.1f} % (2025) → {float(_sh_path.loc[FY[-1]]):.1f} % (2030). Dövlət komponenti artıq "
      f"ayrıca tənliklə idarə olunduğu üçün cəmi ilə birlikdə hərəkət etmir; qeyri-dövlət qalıq "
      f"olduğuna görə bütün fərqi O udur. Bu, icmalın tələb etdiyi bölgünün mexaniki nəticəsidir "
      f"və Bölmə 10-da məhdudiyyət kimi qeyd olunur.")
pd.DataFrame({"dövlət (ansambl) %": state_g, "dövlət səviyyə": state_level,
              "qeyri-dövlət (qalıq)": nonstate_level, "qeyri-dövlət %": nonstate_g,
              "dövlətin payı %": _sh_path}).round(2)
````

### 7.5 İnvestisiya deflyatoru və nominal körpü (investisiyanın ayrıca qiymət indeksi)

Deflyator yolu Bölmə 6.5-in ansamblından gəlir. Sürücülərin yolları: tikinti-quraşdırma
deflyatoru — FR02-nin `pdrv_constr` açarı; idxal qiymətləri — bu dəftərin `import_price_infl`
açarı. Nəticə iki formada nəşr olunur: `inv_defl_g` (illik artım, %) və `inv_defl` (indeks,
**2025 = 100**). Nominal səviyyə yolu artıq sabit qiymət konvensiyası ilə deyil, həqiqi deflyator
ilə qurulur:

```
inv_total_nom(t) = inv_total_real2025(t) × Π_{s≤t} (1 + inv_defl_g(s)/100)
```

````python
_wc3 = float(w_c.iloc[-3:].mean())
_drv_defl = {"tikinti_deflyatoru": {int(y): float(pdrv_constr_fc.loc[y]) for y in FY},
             "idxal_qiymetleri": {int(y): float(imp_fc.loc[y]) for y in FY},
             "iqi": {int(y): float(cpi_fc.loc[y]) for y in FY},
             "proxy_defl": {int(y): float(_wc3 * pdrv_constr_fc.loc[y] + (1 - _wc3) * imp_fc.loc[y])
                    for y in FY},
             "neer_g_l1": {int(y): float(g_neer.loc[LA]) if LA in g_neer.index else 0.0 for y in FY}}


def _recursive_own_lag_path(frame, cols, drivers, years, y_last, lag_col="deflyator_l1"):
    """Öz gecikməsini daşıyan spesifikasiya üçün REKURSİV yol: hər ilin proqnozu növbəti ilin
    gecikmiş dəyəri olur (sabit lövbər qoymaq gizli fərziyyə olardı)."""
    d = frame[["y"] + cols].dropna()
    f = sm.OLS(d["y"], sm.add_constant(d[cols], has_constant="add")).fit()
    out, prev = {}, float(y_last)
    for t in years:
        row = {c: (prev if c == lag_col else float(drivers[c][int(t)])) for c in cols}
        prev = float(f.params["const"] + sum(f.params[c] * row[c] for c in cols))
        out[int(t)] = prev
    return out


defl_paths = {
    "ONLARIN_FORMASI": _recursive_own_lag_path(DF, ["deflyator_l1", "neer_g_l1"], _drv_defl, FY,
                                               float(d_inv_g.loc[LA])),
    "PROXY_QIYMETLI": _lin_path(DF, ["proxy_defl"], _drv_defl, FY),
    "PROXY_TETBIQ": {int(y): float(_drv_defl["proxy_defl"][int(y)]) for y in FY},
    "TIKINTI_IDXAL": _lin_path(DF, ["tikinti_deflyatoru", "idxal_qiymetleri"], _drv_defl, FY),
    "IQI": _lin_path(DF, ["iqi"], _drv_defl, FY),
    "AR1": _ar1_path(d_inv_g, FY),
    "ORTA5": {int(y): float(d_inv_g.dropna().iloc[-5:].mean()) for y in FY},
    "RW": {int(y): float(d_inv_g.dropna().iloc[-1]) for y in FY},
}
inv_defl_g = _combine(bt_defl, defl_paths, FY).rename("inv_defl_g")

# indeks, 2025 = 100
defl_index, _p = {}, 100.0
for y in FY:
    _p = _p * (1 + float(inv_defl_g.loc[y]) / 100.0)
    defl_index[y] = _p
defl_index = pd.Series(defl_index, name="inv_defl")

# nominal körpü
total_nom_fc = (total_level * defl_index / 100.0).rename("inv_total_nom")
dom_nom_fc = (dom_level * defl_index / 100.0).rename("inv_domestic_nom")
for_nom_fc = (for_level * defl_index / 100.0).rename("inv_foreign_nom")
state_nom_fc = (state_level * defl_index / 100.0).rename("inv_state_nom")
nonstate_nom_fc = (nonstate_level * defl_index / 100.0).rename("inv_nonstate_nom")
oil_nom_fc = (oil_level * defl_index / 100.0).rename("inv_oil_nom")
nonoil_nom_fc = (nonoil_level * defl_index / 100.0).rename("inv_nonoil_nom")

print("investisiya deflyatoru — ansambl çəkiləri:", {k: round(v, 3) for k, v in bt_defl.weights.items()})
display(pd.DataFrame({m: pd.Series(p) for m, p in defl_paths.items()}).round(2))
pd.DataFrame({"deflyator artımı %": inv_defl_g, "deflyator indeksi (2025=100)": defl_index,
              "cəmi, 2025 sabit qiy.": total_level, "cəmi, cari qiy. (yeni)": total_nom_fc}).round(2)
````

### 7.6 Yelpik zolaqları

Cəmi investisiya üçün Bölmə 5.1-in strukturuna əsaslanan parametr kovariasiyası və Bölmə 6-nın
`STRUKTUR` OOS RMSE-si istifadə olunur. Xarici investisiya üçün namizəd ansamblının OOS RMSE-si
götürülür. Dövlət və neft-qaz sıraları üçün zolaq öz yarışlarının ansambl RMSE-sindən qurulur.
**Deflyator istisnadır və səbəbi ölçülmüşdür:** onun ansamblının 99,6 %-ni `TIKINTI_IDXAL`
daşıyır, o isə davranış tənliyi deyil, kompozisiya əlaqəsidir (Bölmə 5.5) — sınaqda ona hədəf
ilin FAKTİKİ tikinti deflyatoru verildiyi üçün RMSE 0,08 f.b.-ə enir və zolaq ±0,1 f.b. çıxardı,
halbuki sıranın tarixi σ-sı 6,1 f.b.-dir (2008–2025). Proqnozda həmin giriş FR02-nin `pdrv_constr` yoludur
və öz xətasını daşıyır. Ona görə FR01, Bölmə 7.3-ün presedenti tətbiq olunur: **mərkəzi yol
dəyişmir**, zolağın σ₁-i isə ŞƏRTSİZ etalonun (RW) ölçülmüş xətasından götürülür. Hər halda zolaq AR(1)-uyğun üfüq amili ilə genişlənir. **Qalıq komponentlər üçün ayrıca
zolaq qurulmur** (daxili, qeyri-dövlət, qeyri-neft): onların qeyri-müəyyənliyi iki müstəqil
mənbədən gəlir və hər ikisi öz sırasında artıq göstərilib.

````python
def _rho(s):
    d = pd.DataFrame({"y": s, "l1": pd.Series(s).shift(1)}).dropna()
    return float(sm.OLS(d["y"], sm.add_constant(d[["l1"]], has_constant="add")).fit().params["l1"])


rho_tot, rho_for = _rho(inv_total_g), _rho(inv_for_g)
rho_state, rho_oil, rho_defl = _rho(inv_state_g), _rho(g_oil_inv), _rho(d_inv_g)
print(f"AR(1) rho — cəmi {rho_tot:.3f}, xarici {rho_for:.3f}, dövlət {rho_state:.3f}, "
      f"neft-qaz {rho_oil:.3f}, deflyator {rho_defl:.3f}")

sig1_tot = float(bt.rmse["STRUKTUR"])
sig1_for = float(bt2.ensemble_rmse)
cov_tot = res_inv.cov_params.values

_A6 = outputs.read_assumptions()
if "m2_real_g" not in _A6:
    raise RuntimeError("m2_real_g assumptions.csv-də yoxdur — FR01 icra olunmayıb (DAG).")
M2_PATH = _A6["m2_real_g"].reindex(FY)
dlrm2_fc = {LA: float(dlrm2.loc[LA]), **{y: float(M2_PATH.loc[y]) for y in FY}}
print("real pul kütləsi profili (FR01): " + ", ".join(f"{y}: {float(M2_PATH.loc[y]):.2f}" for y in FY))


def _bands(path, sig1, rho, seed0, cov=None, X=None):
    out = []
    for h, y in enumerate(path.index, start=1):
        sh = sig1 * float(np.sqrt(sum(rho ** (2 * j) for j in range(h))))
        out.append(ek.fan(path.loc[[y]], resid_sigma=sh, coef_cov=cov,
                          X_future=(None if X is None else X[y]), ndraw=6000, seed=ek.SEED + seed0 + h))
    return pd.concat(out)


_X = {y: np.array([[1.0, dlrm2_fc[y], dlrm2_fc[y - 1], 0.0]]) for y in FY}
bands_tot = _bands(total_g, sig1_tot, rho_tot, 0, cov=cov_tot, X=_X)
bands_for = _bands(for_g, sig1_for, rho_for, 100)
bands_state = _bands(state_g, float(bt_state.ensemble_rmse), rho_state, 200)
bands_oil = _bands(oil_g, float(bt_oil_used.ensemble_rmse), rho_oil, 300)
# DÜZƏLİŞ — deflyator zolağının σ₁-i. Əvvəllər bura ansamblın sınaq RMSE-si (0,08 f.b.)
# qoyulurdu və nəticədə nəşr olunan 80 % zolaq ±0,1 f.b. idi — beş illik üfüqdə bir qiymət
# indeksi üçün inandırıcı deyil (sıranın tarixi σ-sı 6,6 f.b., RW etalonunun xətası 6,5 f.b.).
# Səbəb ölçmədədir: ansamblın 99,6 %-ni `TIKINTI_IDXAL` daşıyır, o isə DAVRANIŞ TƏNLİYİ DEYİL —
# Bölmə 5.5-də açıq deyildiyi kimi KOMPOZİSİYA əlaqəsidir (tikinti-quraşdırma deflyatoru cəmi
# investisiya deflyatorunun 60-76 %-lik komponentidir). Sınaqda ona hədəf ilin FAKTİKİ tikinti
# deflyatoru verilir, ona görə o, proqnoz vermir — hədəfi öz komponentindən BƏRPA edir və RMSE
# 0,075-ə enir. Proqnozda isə həmin giriş FR02-nin `pdrv_constr` YOLUDUR və onun öz xətasını
# daşıyır; həmin xəta ölçüdə ümumiyyətlə yoxdur.
# Ona görə FR01, Bölmə 7.3-ün presedenti tətbiq olunur: MƏRKƏZİ YOL DƏYİŞMİR, yalnız zolağın
# eni şərtsiz etalonun (RW) ölçülmüş xətasından götürülür.
SIG1_DEFL_COND = float(bt_defl.ensemble_rmse)          # şərti (kompozisiya) ölçü — istifadə edilmir
SIG1_DEFL = float(bt_defl.rw_rmse)                     # ŞƏRTSİZ etalon — zolağın əsası
print(f"deflyator zolağı: şərti ansambl RMSE {SIG1_DEFL_COND:.3f} f.b. (kompozisiya əlaqəsinin "
      f"bərpa xətası) ƏVƏZİNƏ şərtsiz RW xətası {SIG1_DEFL:.3f} f.b. götürülür; "
      f"sıranın tarixi σ-sı {float(d_inv_g.std(ddof=0)):.2f} f.b. — mərkəzi yol dəyişmir")
bands_defl = _bands(inv_defl_g, SIG1_DEFL, rho_defl, 400)

pd.concat([pd.DataFrame({"cəmi (mərkəzi)": total_g}), bands_tot,
           pd.DataFrame({"dövlət (mərkəzi)": state_g}), bands_state], axis=1).round(2)
````

````python
def _rows(series_code, name_az, unit, actual_s=None, fc_path=None, fc_bands=None,
          fc_source="ours", ministry_s=None):
    """`fc_source` proqnoz sətirlərinin mənbəyini müəyyən edir. Qayda: bir rəqəm ancaq BİR mənbə
    adı altında göndərilir."""
    rows = []
    if actual_s is not None:
        for y, v in actual_s.items():
            if pd.isna(v):
                continue
            rows.append({"fr": "FR6", "series_code": series_code, "series_name_az": name_az, "unit": unit,
                        "year": int(y), "kind": "actual", "source": "ours", "value": float(v),
                        "lo80": None, "hi80": None, "lo50": None, "hi50": None})
    if fc_path is not None:
        for y in fc_path.index:
            b = fc_bands.loc[y] if fc_bands is not None else None
            rows.append({"fr": "FR6", "series_code": series_code, "series_name_az": name_az, "unit": unit,
                        "year": int(y), "kind": "forecast", "source": fc_source, "value": float(fc_path.loc[y]),
                        "lo80": (float(b["lo80"]) if b is not None else None),
                        "hi80": (float(b["hi80"]) if b is not None else None),
                        "lo50": (float(b["lo50"]) if b is not None else None),
                        "hi50": (float(b["hi50"]) if b is not None else None)})
    if ministry_s is not None:
        for y, v in ministry_s.items():
            if pd.isna(v):
                continue
            rows.append({"fr": "FR6", "series_code": series_code, "series_name_az": name_az, "unit": unit,
                        "year": int(y), "kind": "forecast", "source": "template_sample", "value": float(v),
                        "lo80": None, "hi80": None, "lo50": None, "hi50": None})
    return rows


fl_rows = []


def _rows_level(base_code, name_az, actual_s=None, fc_path=None, ministry_s=None, nom_fc=None):
    """Səviyyə sırası İKİ kodla nəşr olunur — qiymət bazası ADIN İÇİNDƏ görünür (D5):

      · `<kod>_nom`       — CARİ qiymətlər: faktiki müşahidələr, şablon-nümunə səviyyələri VƏ
        (bu buraxılışdan e'tibarən) bizim öz nominal proqnozumuz — deflyator yolu ilə qurulub;
      · `<kod>_real2025`  — 2025-ci ilin SABİT qiymətləri: proqnoz yolu + baza ili lövbəri.
    """
    rows = []
    if actual_s is not None:
        for y, v in actual_s.items():
            if pd.isna(v):
                continue
            rows.append({"fr": "FR6", "series_code": f"{base_code}_nom",
                        "series_name_az": f"{name_az} (cari qiymətlərlə)", "unit": "mln AZN",
                        "year": int(y), "kind": "actual", "source": "ours", "value": float(v),
                        "lo80": None, "hi80": None, "lo50": None, "hi50": None})
    if nom_fc is not None:
        for y in nom_fc.index:
            rows.append({"fr": "FR6", "series_code": f"{base_code}_nom",
                        "series_name_az": f"{name_az} (cari qiymətlərlə)", "unit": "mln AZN",
                        "year": int(y), "kind": "forecast", "source": "ours", "value": float(nom_fc.loc[y]),
                        "lo80": None, "hi80": None, "lo50": None, "hi50": None})
    if ministry_s is not None:
        for y, v in ministry_s.items():
            if pd.isna(v):
                continue
            rows.append({"fr": "FR6", "series_code": f"{base_code}_nom",
                        "series_name_az": f"{name_az} (cari qiymətlərlə)", "unit": "mln AZN",
                        "year": int(y), "kind": "forecast", "source": "template_sample",
                        "value": float(v), "lo80": None, "hi80": None, "lo50": None, "hi50": None})
    if fc_path is not None:
        nm_r = f"{name_az} (2025 sabit qiymətləri ilə)"
        if actual_s is not None and LA in actual_s.index and not pd.isna(actual_s.loc[LA]):
            rows.append({"fr": "FR6", "series_code": f"{base_code}_real2025",
                        "series_name_az": nm_r, "unit": "mln AZN", "year": int(LA),
                        "kind": "actual", "source": "ours", "value": float(actual_s.loc[LA]),
                        "lo80": None, "hi80": None, "lo50": None, "hi50": None})
        for y in fc_path.index:
            rows.append({"fr": "FR6", "series_code": f"{base_code}_real2025",
                        "series_name_az": nm_r, "unit": "mln AZN", "year": int(y),
                        "kind": "forecast", "source": "ours", "value": float(fc_path.loc[y]),
                        "lo80": None, "hi80": None, "lo50": None, "hi50": None})
    return rows


fl_rows += _rows_level("inv_total", "Əsas kapitala cəmi investisiya",
                       actual_s=inv_total, fc_path=total_level, ministry_s=mb_total, nom_fc=total_nom_fc)
fl_rows += _rows("inv_total_g", "Əsas kapitala investisiyaların real artım tempi", "%",
                 actual_s=inv_total_g, fc_path=total_g, fc_bands=bands_tot, ministry_s=mb_total_g)
fl_rows += _rows_level("inv_domestic", "Daxili investisiya",
                       actual_s=inv_dom, fc_path=dom_level, ministry_s=mb_dom, nom_fc=dom_nom_fc)
fl_rows += _rows("inv_domestic_g", "Daxili investisiyanın real artım tempi", "%",
                 actual_s=inv_dom_g, fc_path=dom_g)
fl_rows += _rows_level("inv_foreign", "Xarici investisiya",
                       actual_s=inv_for, fc_path=for_level, ministry_s=mb_for, nom_fc=for_nom_fc)
fl_rows += _rows("inv_foreign_g", "Xarici investisiyanın real artım tempi", "%",
                 actual_s=inv_for_g, fc_path=for_g, fc_source="template_sample")
fl_rows += _rows_level("inv_oil", "Neft-qaz sektoruna investisiyalar",
                       actual_s=inv_oil, fc_path=oil_level, nom_fc=oil_nom_fc)
fl_rows += _rows_level("inv_nonoil", "Qeyri neft-qaz sektoruna investisiyalar",
                       actual_s=inv_nonoil, fc_path=nonoil_level, nom_fc=nonoil_nom_fc)
fl_rows += _rows("inv_oil_g", "Neft-qaz sektoruna investisiyaların real artım tempi", "%",
                 actual_s=g_oil_inv, fc_path=oil_g, fc_bands=bands_oil)
fl_rows += _rows("inv_nonoil_g", "Qeyri neft-qaz sektoruna investisiyaların real artım tempi", "%",
                 actual_s=g_nonoil_inv, fc_path=nonoil_g)
fl_rows += _rows_level("inv_state", "Dövlət investisiyaları",
                       actual_s=inv_state, fc_path=state_level, nom_fc=state_nom_fc,
                       ministry_s=data_layer.ministry_baseline("inv_state"))
fl_rows += _rows("inv_state_g", "Dövlət investisiyalarının real artım tempi", "%",
                 actual_s=inv_state_g, fc_path=state_g, fc_bands=bands_state)
fl_rows += _rows_level("inv_nonstate", "Qeyri-dövlət investisiyaları",
                       actual_s=inv_nonst, fc_path=nonstate_level, nom_fc=nonstate_nom_fc,
                       ministry_s=data_layer.ministry_baseline("inv_nonstate"))
fl_rows += _rows("inv_nonstate_g", "Qeyri-dövlət investisiyalarının real artım tempi", "%",
                 actual_s=inv_nonst_g, fc_path=nonstate_g)

# --- YENİ sıralar: qeyri-neft × mülkiyyət, deflyator, fiskal blok --------------------------------
nn_nom_fc = (nn_level * defl_index / 100.0)
nnst_nom_fc = (nnst_level * defl_index / 100.0)
fl_rows += _rows_level("inv_nonoil_nonstate", "Qeyri neft-qaz qeyri-dövlət investisiyaları",
                       actual_s=nonoil_nonstate, fc_path=nn_level, nom_fc=nn_nom_fc)
fl_rows += _rows_level("inv_nonoil_state", "Qeyri neft-qaz dövlət investisiyaları",
                       actual_s=nonoil_state, fc_path=nnst_level, nom_fc=nnst_nom_fc)
fl_rows += _rows("inv_nonoil_nonstate_g", "Qeyri neft-qaz qeyri-dövlət investisiyalarının real artımı",
                 "%", actual_s=NN["y"].dropna(), fc_path=nn_g)
# Dövlət ayağının ARTIM sırası da nəşr olunur (icmalın B10 qeydi): əvvəllər bu ayaq yalnız
# səviyyə (cari və 2025 sabit qiymətlər) kimi verilirdi, qeyri-dövlət ayağı isə artım tempi ilə —
# yəni bölgü ekranda simmetrik deyildi. Sıra EYNİLİK qalığıdır (qeyri-neft cəmi − qeyri-dövlət),
# ona görə öz tənliyi yoxdur; bu, sıra lüğətində və FR6 səhifəsində açıq yazılır.
fl_rows += _rows("inv_nonoil_state_g", "Qeyri neft-qaz dövlət investisiyalarının real artımı",
                 "%", actual_s=real_g(nonoil_state), fc_path=nnst_g)
fl_rows += _rows("inv_defl_g", "İnvestisiya deflyatorunun illik artımı", "%",
                 actual_s=d_inv_g, fc_path=inv_defl_g, fc_bands=bands_defl)
fl_rows += _rows("inv_defl", "İnvestisiya deflyatoru (indeks, 2025 = 100)", "indeks",
                 actual_s=pd.Series({LA: 100.0}), fc_path=defl_index)
# DƏVQ: faktiki (cari qiymətlər) + proqnoz yolu (cari qiymətlər, real artım × İQİ)
capex_nom_fc, _p = {}, float(capex.loc[LA])
for _y in FY:
    _p = _p * (1 + float(capex_realg.loc[_y]) / 100.0) * (1 + float(cpi_fc.loc[_y]) / 100.0)
    capex_nom_fc[_y] = _p
capex_nom_fc = pd.Series(capex_nom_fc, name="state_capex")
fl_rows += _rows("state_capex", "Dövlət büdcəsinin əsaslı vəsait qoyuluşu (DƏVQ)", "mln AZN",
                 actual_s=capex, fc_path=capex_nom_fc)
# --- FİSKAL PARAMETRLƏR VƏ QEYRİ-NEFT BAZA KƏSİRİNİN ÜFÜQƏ UZADILMASI (icmalın B18 qeydi) ------
# Qeyd üç şey istəyir: (i) fiskal parametrlərin görünməsi, (ii) xərclərin cari / əsaslı / borca
# xidmət ayrılığı, (iii) qeyri-neft baza kəsirinin qeyri-neft ÜDM-ə nisbətinin modelə daxil olması.
# (i) və (ii) məlumat qatında ONSUZ DA var idi, lakin `forecast_long`-a YAZILMIRDI — ona görə
# oxucu "fiskal məlumat yoxdur" nəticəsinə gəlirdi. İndi hamısı nəşr olunur.
# (iii) sıra yalnız TARİXDƏ mövcud idi (2010–2025); aşağıda üfüqə uzadılır.
#
# UZATMA QAYDASI — TƏNLİK DEYİL, KALİBRLƏNMİŞ PAY QAYDASIDIR və belə də etiketlənir:
# qeyri-neft büdcə gəlirləri və büdcə xərcləri son beş faktiki ilin qeyri-neft ÜDM-ə nisbətində
# saxlanılır, səviyyələr isə FR02-nin nəşr etdiyi NOMİNAL qeyri-neft ÜDM yolundan qurulur.
# Bu, siyasət proqnozu deyil — DƏYİŞMƏYƏN SİYASƏT PAYLARI altında baza yoludur; nisbətin özü
# demək olar ki, sabit çıxır və bu, qaydanın nəticəsidir, tapıntı deyil.
_nn_nom = (pd.read_csv(os.path.join(config.OUT, "forecast_long.csv"))
             .query("series_code == 'nonoil_nom' and source == 'ours'")
             .set_index("year")["value"].astype(float).sort_index())
# LÖVBƏR SON FAKTİKİ İLDİR, beş illik orta DEYİL. Səbəb ölçülüb: beş illik orta paylarla
# nisbət 2025-in faktiki −19,77 %-indən 2026-da −22,92 %-ə SIÇRAYIR, yəni 2025/2026 sərhədində
# süni qırılma yaranır — paketin D5 qaydası məhz bu növ qırılmanı qadağan edir. Son ilin payları
# ilə yol kəsilmədən davam edir. Beş illik orta variantı SİLİNMİR, aşağıda həssaslıq kimi çap olunur.
_SHARE_WIN = 5
_rev_share = float((nob["revenue_nonoil"] / nob["gdp_nonoil"]).loc[LA])
_exp_share = float((nob["expenditure"] / nob["gdp_nonoil"]).loc[LA])
_rev_share5 = float((nob["revenue_nonoil"] / nob["gdp_nonoil"]).loc[LA - _SHARE_WIN + 1:LA].mean())
_exp_share5 = float((nob["expenditure"] / nob["gdp_nonoil"]).loc[LA - _SHARE_WIN + 1:LA].mean())
_rev_fc = (_nn_nom.reindex(FY) * _rev_share).rename("revenue_nonoil")
_exp_fc = (_nn_nom.reindex(FY) * _exp_share).rename("expenditure")
_bal_fc = (_rev_fc - _exp_fc).rename("balance")
_ratio_fc = (_bal_fc / _nn_nom.reindex(FY) * 100.0).rename("ratio")
print(f"\nB18 — qeyri-neft baza kəsirinin üfüqə uzadılması (pay qaydası, lövbər {LA}):")
print(f"  qeyri-neft gəlirlərinin payı {_rev_share*100:.2f} % · büdcə xərclərinin payı "
      f"{_exp_share*100:.2f} % (qeyri-neft ÜDM-ə nisbətdə)")
print(pd.DataFrame({"qeyri-neft gəlir": _rev_fc.round(1), "büdcə xərcləri": _exp_fc.round(1),
                    "baza balansı": _bal_fc.round(1), "nisbət, %": _ratio_fc.round(2)}).to_string())
print(f"  faktiki {LA} nisbəti: {float(nob['ratio_nonoil_gdp'].loc[LA]):.2f} % — "
      f"yol kəsilmədən davam edir (qırılma yoxdur)")
print(f"  HƏSSASLIQ: beş illik orta paylarla ({_rev_share5*100:.2f} % / {_exp_share5*100:.2f} %) "
      f"nisbət {(_rev_share5 - _exp_share5) * 100:.2f} % olardı — {LA}-in faktiki dəyərindən "
      f"{abs((_rev_share5 - _exp_share5) * 100 - float(nob['ratio_nonoil_gdp'].loc[LA])):.2f} f.b. "
      "aralı; həmin variant süni sıçrayış yaratdığı üçün nəşr olunmur.")

# fiskal parametrlər — tarixi sıralar nəşr olunur (proqnoz yolu yalnız yuxarıdakı üç sıra üçün)
for _code, _nm, _col in [
        ("budget_revenue_total", "Dövlət büdcəsinin gəlirləri", "revenue_total"),
        ("budget_revenue_oil", "Dövlət büdcəsinin neft-qaz gəlirləri", "revenue_oil"),
        ("budget_revenue_nonoil", "Dövlət büdcəsinin qeyri-neft-qaz gəlirləri", "revenue_nonoil"),
        ("budget_expenditure", "Dövlət büdcəsinin xərcləri", "exp_total"),
        ("budget_exp_current", "Dövlət büdcəsinin cari xərcləri", "exp_current"),
        ("budget_exp_capital", "Dövlət büdcəsinin əsaslı xərcləri", "exp_capital"),
        ("budget_debt_service", "Borca xidmət xərcləri", "debt_service"),
        ("budget_balance", "Dövlət büdcəsinin kəsiri/profisiti", "balance")]:
    if _col in fisc:
        fl_rows += _rows(_code, _nm, "mln AZN", actual_s=fisc[_col].dropna())

fl_rows += _rows("nonoil_base_balance_sb", "Dövlət büdcəsi üzrə qeyri-neft balansı", "mln AZN",
                 actual_s=nob["balance"], fc_path=_bal_fc)
fl_rows += _rows("nonoil_base_ratio_sb",
                 "Dövlət büdcəsi üzrə qeyri-neft balansı / qeyri-neft ÜDM", "%",
                 actual_s=nob["ratio_nonoil_gdp"], fc_path=_ratio_fc)

fl_df = pd.DataFrame(fl_rows)
_dup = (fl_df[fl_df.kind == "forecast"]
        .groupby(["series_code", "year"])["value"]
        .agg(lambda s: len(s) > 1 and s.round(6).nunique() == 1))
assert not _dup.any(), f"bit-bərabər dublikat mənbə: {sorted(_dup[_dup].index.tolist())}"
for _c in ("inv_total", "inv_domestic", "inv_foreign", "inv_oil", "inv_nonoil",
           "inv_state", "inv_nonstate", "inv_nonoil_nonstate", "inv_nonoil_state"):
    _n = fl_df[(fl_df.series_code == f"{_c}_nom") & (fl_df.year == LA) & (fl_df.kind == "actual")]
    _r = fl_df[(fl_df.series_code == f"{_c}_real2025") & (fl_df.year == LA)]
    assert len(_n) == 1 and len(_r) == 1 and abs(float(_n.value.iloc[0]) - float(_r.value.iloc[0])) < 1e-9, \
        f"{_c}: baza ilində (2025) cari və sabit qiymətli səviyyələr üst-üstə düşmür"
outputs.write_forecast_long(fl_df)
print(f"forecast_long.csv-ə {len(fl_df)} sətir yazıldı (FR6 əhatəsi); "
      f"o cümlədən şablon-nümunə mənbəli: {int((fl_df['source'] == 'template_sample').sum())}")
fl_df.groupby(["series_code", "kind"]).size().unstack(fill_value=0)
````

````python
# Bir model ümumi pəncərədə minimum nöqtə sayını doldura bilmirsə, onun RMSE-si təyin olunmur
# (None) — belə sətirlər çıxış müqaviləsinə yazılmır, lakin NEÇƏSİNİN atıldığı çap olunur ki,
# "sınandı, lakin ölçülə bilmədi" halı görünməz qalmasın.
def _wbt(res, label):
    t = res.table
    drop = int(t["rmse_h"].isna().sum())
    if drop:
        print(f"  {label}: {drop} sətir yazılmadı — modelin RMSE-si ümumi pəncərədə təyin olunmur "
              f"(minimum nöqtə sayı dolmayıb)")
    outputs.write_backtest(t.dropna(subset=["rmse_h"]))


_wbt(bt, "inv_total_g")
_wbt(bt2, "inv_foreign_g")
_wbt(bt_state, "inv_state_g")
_wbt(bt_oil, "inv_oil_g")
_wbt(bt_oil_used, "inv_oil_g_merkezi")
_wbt(bt_defl, "inv_defl_g")
_wbt(bt_nn, "inv_nonoil_nonstate_g")
print("validation_backtest.csv: FR6-nın yeddi yarışı yazıldı "
      "(inv_total_g, inv_foreign_g, inv_state_g, inv_oil_g, inv_oil_g_merkezi, inv_defl_g, "
      "inv_nonoil_nonstate_g)")
````

````python
v = validate.Verifier("FR06 — İnvestisiyalar və fiskal blok")

# 0. ƏSAS UYĞUNLUQ (§D3): cəmi investisiya yolu = FR01 nüvəsinin yolu, HƏR il
core_path_level = {}
_p = float(inv_total.loc[LA])
for _y in FY:
    _p = _p * (1 + float(g_inv_core.loc[_y]) / 100.0)
    core_path_level[_y] = _p
core_path_level = pd.Series(core_path_level)
validate.identity_check("FR06 cəmi investisiya = FR01 nüvəsinin yolu (səviyyə, hər il)",
                        total_level, core_path_level, tol=1e-9, verbose=True)
v.identity("FR06 cəmi investisiya = FR01 nüvəsinin yolu (səviyyə, hər il)", total_level, core_path_level, tol=1e-9)
v.identity("FR06 cəmi artım tempi = g_inv_core (hər il)", total_g, g_inv_core.loc[FY], tol=1e-12)

# 1-3. proqnoz eynilikləri — HƏR il
v.identity("cəmi = daxili + xarici (2026-2030)", total_level, dom_level + for_level, tol=1e-9)
v.identity("xarici artım tempi = inv_foreign_realg fərziyyəsi", for_g, fdi_realg.loc[FY], tol=1e-12)
v.identity("cəmi = dövlət + qeyri-dövlət (2026-2030)", total_level, state_level + nonstate_level,
           tol=1e-6, rel_tol=1e-9)
v.identity("cəmi = neft-qaz + qeyri-neft-qaz (2026-2030)", total_level, oil_level + nonoil_level,
           tol=1e-6, rel_tol=1e-9)
v.identity("qeyri-neft = qeyri-neft dövlət + qeyri-neft qeyri-dövlət (2026-2030)",
           nonoil_level, nnst_level + nn_level, tol=1e-6, rel_tol=1e-9)

# 4. tarixi eyniliklər — HƏR il, 2006-2025 pəncərəsi
v.identity("tarixi: cəmi = daxili + xarici (2006-2025)", inv_total.loc[2006:2025],
           (inv_dom + inv_for).loc[2006:2025], tol=1e-6, rel_tol=1e-5)
v.identity("tarixi: cəmi = dövlət + qeyri-dövlət (2006-2025)", inv_total.loc[2006:2025],
           (inv_state + inv_nonst).loc[2006:2025], tol=1e-6, rel_tol=1e-5)
v.identity("tarixi: cəmi = neft-qaz + qeyri-neft-qaz (2006-2025)", inv_total.loc[2006:2025],
           (inv_oil + inv_nonoil).loc[2006:2025], tol=1e-6, rel_tol=1e-5)
v.identity("tarixi: qeyri-neft = dövlət + qeyri-dövlət (2018-2025)",
           nonoil_inv_wb.loc[2018:LA], (nonoil_state + nonoil_nonstate).loc[2018:LA], tol=1e-6)

# 5. nominal körpü eyniliyi: nominal = real × deflyator indeksi, HƏR il
v.identity("nominal körpü: inv_total_nom = inv_total_real2025 × deflyator indeksi",
           total_nom_fc, total_level * defl_index / 100.0, tol=1e-9)

# 6. fiskal blok eynilikləri — HƏR il
v.identity("qeyri-neft balansı = qeyri-neft gəlirlər − xərclər (hər il)",
           nob["balance"], nob["revenue_nonoil"] - nob["expenditure"], tol=1e-9)
v.identity("qeyri-neft nisbəti = balans / qeyri-neft ÜDM × 100 (hər il)",
           nob["ratio_nonoil_gdp"], nob["balance"] / nob["gdp_nonoil"] * 100.0, tol=1e-9)
# 6b. İKİ BÖLGÜ OXUNUN UZLAŞMASI: mülkiyyət və sektor bölgüləri müstəqil oxlardır, lakin
#     onların kəsişməsi mənfi ola bilməz — neft-qaz sektorundakı dövlət investisiyası
#     (cəmi dövlət − qeyri-neft dövlət) həm müsbət, həm də cəmi neft-qaz investisiyasından
#     kiçik olmalıdır. Bu, iki ayrı ansamblın bir-birindən ayrı düşmədiyinin yeganə yoxlamasıdır.
_oil_state_implied = (state_level - nnst_level)
v.ok("kəsişmə: neft-qaz sektorunda dövlət investisiyası ≥ 0 (hər il)",
     bool((_oil_state_implied.values >= -1e-9).all()),
     detail=f"min {float(_oil_state_implied.min()):.1f} mln AZN")
v.ok("kəsişmə: neft-qaz sektorunda dövlət investisiyası ≤ cəmi neft-qaz investisiyası (hər il)",
     bool((_oil_state_implied.values <= oil_level.values + 1e-9).all()),
     detail=f"maks. pay {float((_oil_state_implied / oil_level).max() * 100):.1f} %")

v.ok("DƏVQ birləşməsi 2022-2023 illərini doldurur",
     bool({2022, 2023} <= set(capex.index)))
v.ok("DƏVQ 2025 = fiscal_r019 2025 (birləşmə mənbəni əvəz etmir)",
     abs(float(capex.loc[LA]) - float(fisc["state_capex_fiscal"].loc[LA])) < 1e-9)

# 7. müstəqil yenidən hesablamalar
v.recompute("2026 xarici investisiya səviyyəsi (əl ilə)", shipped=float(for_level.loc[2026]),
            fn=lambda: float(inv_for.loc[LA]) * (1 + float(fdi_realg.loc[2026]) / 100.0), tol=1e-6)
# dözüm 1e-4: `assumptions.csv` dəyərləri dörd onluq rəqəmə yuvarlaqlaşdırılır (bütün açarlarda
# eyni konvensiya), ona görə geri oxunmuş dəyər ilə tam dəqiqlikli yenidən hesablama arasındakı
# fərq yuvarlaqlaşdırmanın özüdür.
v.recompute("2026 DƏVQ real artımı = DİP planından (əl ilə)",
            shipped=float(capex_realg.loc[FY[0]]),
            fn=lambda: ((dip_plan_2026 / float(capex.loc[LA])) /
                        (1 + float(cpi_fc.loc[FY[0]]) / 100.0) - 1) * 100, tol=1e-4)
v.recompute("2030 nominal cəmi investisiya (əl ilə)", shipped=float(total_nom_fc.loc[2030]),
            fn=lambda: float(total_level.loc[2030]) * float(defl_index.loc[2030]) / 100.0, tol=1e-9)

# 8. yelpik zolaqları mərkəzi yolu əhatə edir
for _nm, _b, _p in [("cəmi", bands_tot, total_g), ("xarici", bands_for, for_g),
                    ("dövlət", bands_state, state_g), ("neft-qaz", bands_oil, oil_g),
                    ("deflyator", bands_defl, inv_defl_g)]:
    v.ok(f"{_nm}: 80% zolağı mərkəzi yolu əhatə edir",
         bool(((_b["lo80"] < _p.values) & (_b["hi80"] > _p.values)).all()))

# 9. gələcəyə baxış yoxdur
for nm, idx in [("kredit kanalı (STRUKTUR)", sI.index), ("kredit kanalı (dummisiz)", sIu.index),
                ("kredit kanalı (2015-16 çıxarılıb)", sIx.index), ("dövlət payı trendi", sh_state.index),
                ("dövlət investisiyası tənliyi", _s2.index), ("neft investisiyası", OI.dropna(subset=["y"]).index),
                ("investisiya deflyatoru", DF.dropna(subset=["y"]).index),
                ("qeyri-neft mülkiyyət bölgüsü", NN.dropna(subset=["y"]).index)]:
    v.no_lookahead(nm, idx)

v.assert_all()
````

## 10. Məhdudiyyətlər

**Şablon-nümunə sütunu iki sırada 2026-da bitir.** Statutar şablonun 2.4.1.5.-7. vərəqində
investisiyanın dövlət/qeyri-dövlət bölgüsünü daşıyan sətir (r23, «Cəmi investisiyalarda xüsusi
çəkisi») proqnoz blokunda YALNIZ 2025 və 2026 sütunlarını doldurub; 2027–2030 boşdur. Ona görə
`inv_state_nom` və `inv_nonstate_nom` kodlarının **`source=template_sample`** sətirləri yalnız
2026-cı il üçün mövcuddur. Bu, paketin proqnozunda boşluq DEYİL — paketin öz yolu
(`source=ours`) hər beş il üçün doludur; boşluq müqayisə mənbəyinin özündədir və doldurulmur,
çünki Nazirliyin şablonunda olmayan rəqəmi uydurmaq olmaz. Ekranda müqayisə sütunu həmin sıralar
üçün 2027–2030-da boş görünəcək və bu, gözlənilən davranışdır.

**DATA-GAP: 2000–2005 investisiya buraxılış fərqi.** `inv_total` (Real sektor sətir 80) 2000–2005
illərində üç bölgünün hər birinin cəmindən eyni ədədlə fərqlənir; komponent sıraları həmin illər
üçün daha məhdud bir buraxılışa aiddir. 2006-cı ildən fərq praktik olaraq sıfırdır. Tarixi eynilik
yoxlamaları 2006–2025 pəncərəsi ilə aparılır, 2000–2005 sətirləri isə `forecast_long.csv`-də
faktiki olaraq qalır (məlumat gizlədilmir). 2025-ci ildə ~0,02 mln AZN qalıq fərq var (rəsmi məcmu
göstərici komponentlərlə tam üst-üstə düşmür), ona görə `rel_tol=1e-5`.

**İki rəsmi mənbə DƏVQ üçün fərqli rəqəm verir.** `fiscal_r019` və `DİP 2016-2026` r4 2016–2020 və
2025-də üst-üstə düşür, 2021-də **46,4 %**, 2024-də **13,7 %** fərqlənir. Birləşdirmə yalnız
`fiscal_r019`-un boş illərinə (2022, 2023) tətbiq olunur; fərq düzəldilmir və Bölmə 2.2-də cədvəl
şəklində nəşr olunur. Hansı sıranın hansı anlayışı ölçdüyünün təsdiqi məlumat sorğusundadır.

**dövlət investisiyası tənliyinin ikinci termi artıq qiymətləndirilir — və nəticə mənfidir.** İki termli forma ikinci term üçün `real_r114`-ü göstərir; əvvəlki buraxılışda həmin sətrin illik sütunu yalnız 2023–2025 üçün dolu idi (n = 3) və term qiymətləndirilə bilmirdi. Bu buraxılışda eyni anlayışın 1999–2024 tarixçəsi açıq mənbədən (opendata.az, maliyyə mənbələri üzrə investisiyalar) əlavə edilib — üst-üstə düşən iki ildə iki sıra **dəqiq eynidir**, ona görə birləşdirmə miqyaslama tələb etmir. Tənlik quruldu və qiymətləndirildi: iki termli formada əmsal **sıfırdan ayırd edilə bilmir** (β₂ ≈ +0,02, p ≈ 0,95), düzəldilmiş R² tək termli formadan aşağı düşür, genişlənən pəncərəli sınaqda isə forma təsadüfi gəzişməni **üstələmir**. Tam 26 illik nümunədə tək başına da əlaqə yoxdur (β = −0,10, p = 0,73). Səbəb məlumatın azlığı deyil, **anlayış fərqidir**: sıra maliyyələşmə mənbəyi üzrə bölgüdür (bütün müəssisələrin öz vəsaitləri), mülkiyyət üzrə deyil. Nəticə hər üç yerdə — tənlik kataloqunda, sınaq cədvəlində və burada — olduğu kimi qeyd olunur; mərkəzi yola isə yalnız qapıdan və sınaqdan keçən modellər daxil olur.

**Dövlət investisiyası yarışının nəticəsi düzəldilmir.** Geriyə doğru sınaqda v2-nin üç illik pay
qaydası dövlət investisiyası tənliyindən daha yaxşı nəticə verir; ona görə mərkəzi yolda hər ikisi iştirak edir və
çəkiləri sınaqdan gəlir, iqtisadi mülahizədən yox. Rəyin tələb etdiyi kanal (büdcə capex) beləliklə
modelə DAXİL OLUR, lakin ona sınağın verdiyindən artıq çəki verilmir.

**Dövlət komponentinin ayrıca tənliyi payın SÜRÜŞMƏSİNƏ gətirir.** Dövlət investisiyası artıq
cəmi yolun sabit payı deyil, öz ansamblının yoludur; qeyri-dövlət isə qalıqdır. Nəticədə dövlətin
cəmi investisiyadakı payı üfüq boyu artır (rəqəm Bölmə 7.2-də çap olunur) və bütün fərqi qeyri-dövlət
komponenti udur. Bu, icmalın tələb etdiyi bölgünün mexaniki nəticəsidir və hamarlaşdırılmır: qalığa
süni düzəliş vermək iki şəffaf girişdən birini gizlədərdi. Oxunuşda nəzərə alınmalıdır ki,
qeyri-dövlət investisiyasının illik profili müstəqil bir qiymətləndirmə deyil, iki verilmiş yolun
fərqidir. İki bölgü oxunun (mülkiyyət və sektor) kəsişməsi Bölmə 9-da yoxlanılır: neft-qaz
sektorundakı dövlət investisiyası hər il həm müsbət, həm də cəmi neft-qaz investisiyasından kiçik
qalmalıdır.

**Neft-qaz investisiyasında struktur sürücü tapılmadı.** Neft qiyməti, çəkili qiymət indeksi və
əmtəəlik həcm indeksinin əmsalları hamısı MƏNFİ və (biri istisna olmaqla) əhəmiyyətsizdir — yəni
gözlənilən iqtisadi işarəyə ziddir. Post-hoc izah verilmir; qapı bağlanır və mərkəzi yol dinamik
etalonlara keçir. Eyni yarışda sabit 27,07 %-lik pay qaydası **təsadüfi gəzişmədən də pis** nəticə
verir, ona görə o da mərkəzi yolu daşımır. Nəticə: bölgü artıq sabit pay deyil, lakin onu daşıyan
da struktur tənlik deyil — bu, məlumatın verdiyi cavabdır.

**İnvestisiya deflyatoru davranış tənliyi deyil, kompozisiya əlaqəsidir.** Qazanan spesifikasiyanın
sürücüsü (tikinti-quraşdırma işlərinin implisit deflyatoru) cəmi investisiya deflyatorunun
60–76 %-lik komponentidir; düzəldilmiş R²-nin 1-ə yaxın olması buna görədir və tapıntı kimi təqdim
edilmir. **Bunun iki ölçmə nəticəsi var və hər ikisi açıq deyilməlidir.** Birincisi, sınaq
cədvəlindəki ~99 %-lik "bacarıq" rəqəmi proqnoz bacarığı deyil: sınaqda modelə hədəf ilin faktiki
tikinti deflyatoru verilir, yəni hədəf öz komponentindən bərpa olunur (RMSE 0,08 f.b.). İkincisi —
və bu buraxılışda **düzəldilib** — həmin rəqəm əvvəllər yelpik zolağının da əsası idi və nəşr
olunan 80 % zolaq ±0,1 f.b. çıxırdı, halbuki sıranın tarixi σ-sı 6,1 f.b.-dir (2008–2025). Proqnozda giriş
FR02-nin `pdrv_constr` yoludur və öz xətasını daşıyır; həmin xəta ölçüdə yoxdur. İndi zolağın σ₁-i
şərtsiz RW etalonundan götürülür (mərkəzi yol dəyişmir) — FR01, Bölmə 7.3-ün eyni presedenti. Yolun özü isə FR02-nin `pdrv_constr` yolundan gəlir, yəni deflyatorun qeyri-müəyyənliyi
FR02-nin tikinti qiymət yolunun qeyri-müəyyənliyidir. Nazirliyin `eq_defid` tənliyi qəbul
edilib, bizim qatda yenidən qiymətləndirilib və sınaqda RW-ni üstələmədiyi olduğu kimi nəşr olunur.

**İnvestisiya deflyatorunun 2016 müşahidəsi mənbələr arasında ziddiyyətlidir.** Bizim qatın
implisit deflyatoru 2016-cı ildə +26,24 % verir (Nazirliyin öz CƏMİ deflyator sətri ilə eyni:
26,23964), onların `DEFID` (yalnız DAXİLİ investisiya) sırası isə +2,26 %. Fərq tərif fərqidir:
nümunə faylı devalvasiyanın qiymət effektini xarici investisiya ayağına yükləyir. Düzəliş
aparılmır; hər iki sıra çap olunur.

**Qeyri-neft × mülkiyyət bölgüsü səkkiz müşahidəlidir.** `real_r102` 2018-ci ildən mövcuddur, sınaq
pəncərəsi isə üç-dörd ildir. Nəticə qərar üçün istifadə olunur, lakin nümunə ölçüsü cədvəldə,
kataloqda və seriya lüğətində açıq göstərilir; 2010–2017 üçün geriyə tarixçə məlumat sorğusundadır.

**Qeyri-neft balansı göstəricisi İCMAL BÜDCƏ DEYİL.** Nəşr olunan sıra dövlət büdcəsi analoqudur
(`fiscal_r008 − fiscal_r016`), çünki icmal perimetri (DSMF, İSF, İTSF, Naxçıvan MR) əsas iş
kitabında ümumiyyətlə yoxdur və rəsmi "baza" düzəliş qaydası bizdə deyil. Nazirliyin nümunə
faylındakı icmal sırası yalnız **çarpaz yoxlama** kimi çap olunur (2015–2024; həmin faylın 41-ci
sətrinin məxrəci 2015-ə qədər sınıqdır) və heç bir çıxış faylına yazılmır. İki perimetr eyni
böyüklük sırasındadır (2024: −21,0 % və −20,1 %), lakin eyni deyil — ona görə etiket bunu deyir.

**Büdcənin gələcək yolu üçün rəsmi mənbə yoxdur.** İş kitabında yeganə irəliyə baxan xana DİP-in
2026 "Nəzərdə tutulmuş vəsait" rəqəmidir (2 700,0 mln AZN). 2027–2030 üçün DƏVQ yolu sıranın öz
tarixçəsinə kalibrlənmiş profil sınağından gəlir və bu, `state_capex_realg` açarının qeydində
yazılıb. Rəsmi orta müddətli büdcə çərçivəsi daxil olduqda bir açar dəyişməklə bütün dövlət
investisiyası yolu yenilənir.

**Xarici investisiya ekzogen fərziyyədir, model deyil.** Mərkəzi yol Nazirliyin FDI baza
ssenarisinin real artım tempi ilə eynidir. Nazirliyin nominal səviyyə sütunu istifadə edilmir:
implisit 2025 bazası (5 222,6 mln AZN) iş kitabının dəyərindən (4 203,0 mln AZN) 24,3 % yuxarıdır.

**Nüvənin investisiya yolu real pul kütləsi fərziyyəsinə həssasdır — İŞARƏ səviyyəsində.**
`g_inv_core` FR01-in kredit kanalı tənliyindən gəlir; tənliyin sabit vəziyyətdəki sıfır nöqtəsi
7,48 %-dir və elan olunmuş həssaslıq aralığının (3,23–9,24 %) daxilindədir. Yəni enən investisiya
yolu bir fərziyyənin nəticəsidir, məlumatın məcburi nəticəsi deyil.
