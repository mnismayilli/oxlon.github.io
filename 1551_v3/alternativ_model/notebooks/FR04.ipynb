### 2.1. `#REF!`-ölü 11 qida məhsulunun dirçəldilməsi (bu buraxılış)

Bu buraxılışa qədər 48–58-ci sətirlərin hər biri yalnız beş real il (2013–2017) daşıyırdı və `n_eff<8` qaydası onları avtomatik **SABİT** (2017-ci ilin dəyərində düz xətt) kateqoriyasına yönləndirirdi — yəni on bir qida məhsulunun 2018-dən sonrakı bütün dinamikası modeldən kənarda qalırdı.

`DATA_GAP_NEGATIVE_PROOF.md` sənədinin (vii) bəndi bu on bir sətrin **DSK qarşılıqlarını xana səviyyəsində** təsbit edib: sənaye məhsullarının natural ifadədə istehsalı üzrə **cədvəl 18** (`018en.xls`, 1995–2024) və ət/süd üçün kənd təsərrüfatı cədvəlləri **1.46 / 1.47** (`1.46en.xls`, `1.47en.xls`, 1985–2024). Sıralar `data_layer` §14-ün dondurulmuş açıq-mənbə panelindədir və `vereq_product_concordance()` hər şablon sətrini öz DSK mənbəyi, mənbənin etiketi, vahidi və çevirmə əmsalı ilə birlikdə qaytarır.

**Birləşdirmə qaydası (nə üçün məhz belə).** DSK sırası şablonun sətri ilə **eyni anlayış deyil** (məsələn 1.47 xam süd istehsalını verir, şablon isə emal edilmiş duru südü); ona görə DSK-dan **səviyyə deyil, dinamika** götürülür və sıra şablonun öz son ortaq ilinin (**2017**) səviyyəsinə **yenidən bazalaşdırılır**:

$$Q^{\text{yeni}}_t = Q^{\text{şablon}}_{2017}\times\frac{S^{\text{DSK}}_t}{S^{\text{DSK}}_{2017}}$$

Bu, iki qüsuru eyni anda bağlayır: ölçü vahidi fərqi (ton → min ton) və anlayış fərqi (xam/emal edilmiş) — hər ikisi sabit əmsal olduqda tam, dəyişkən olduqda isə ölçülə bilən qalıqla. Ona görə **hər sətir üçün ortaq illərdə (2013–2017) nisbətin sabitliyi yoxlanılır** və qərar bu yoxlamanın nəticəsinə görə verilir: nisbətin variasiya əmsalı (CV) 5 %-dən böyükdürsə, sıra dirçəldilmir və SABİT qalır.

```python
# ---- konkordans və birləşdirmə diaqnostikası -------------------------------------------
CON = data_layer.vereq_product_concordance()
REBASE_ANCHOR = 2017          # şablonun son real ili — yenidən bazalaşdırma lövbəri
RATIO_CV_MAX = 5.0            # QAPI-2a: nisbətin sabitlik həddi, %
ANCHOR_ID_TOL = 1.0           # QAPI-2b: lövbər ilində eynilik həddi, %

REBASED, revive_rows = {}, []
for _, cr in CON.iterrows():
    _row = int(cr["vereq_row"])
    _Qt = data_layer.vereq_actuals("2.4.1.4.", cr["vereq_name_az"], row=_row)
    _S = data_layer.public_series(cr["series_code"])
    _com = sorted(set(_Qt.index) & set(_S.index))
    _rat = (_Qt.loc[_com] / _S.loc[_com]) if _com else pd.Series(dtype=float)
    _cv = float(_rat.std() / _rat.mean() * 100) if len(_rat) > 1 else np.nan
    _has_anchor = (REBASE_ANCHOR in _Qt.index) and (REBASE_ANCHOR in _S.index)
    _anch = float(_Qt.loc[REBASE_ANCHOR] / _S.loc[REBASE_ANCHOR]) if _has_anchor else np.nan

    # QAPI-1: ölçü vahidi uzlaşırmı (konkordansın xana səviyyəsində yoxlanmış `factor` sahəsi)
    _g1 = bool(pd.notna(cr["factor"]))
    # QAPI-2a: mütənasiblik — ortaq illərdə nisbətin variasiyası kiçikdir
    _g2a = bool(len(_rat) > 1 and _cv <= RATIO_CV_MAX)
    # QAPI-2b: lövbər eyniliyi — lövbər ilində iki sıra EYNİ ədəddir (şablonun ÖZ tərif
    # sınığı ortaq pəncərəni pozduqda 2a keçmir, lakin lövbərdə eynilik dirçəlişi əsaslandırır)
    _g2b = bool(_has_anchor and abs(_anch - 1.0) * 100 <= ANCHOR_ID_TOL)
    _ok = bool(_g1 and _has_anchor and (_g2a or _g2b))
    if _ok:
        REBASED[_row] = (_S * _anch).rename(f"v4_r{_row:03d}_rebased")
    revive_rows.append(dict(
        sətir=_row, məhsul=cr["vereq_name_az"], vahid=cr["vereq_unit"],
        DSK_mənbə=cr["source_file"].split("/")[-1],
        DSK_əhatə=f"{int(cr['year_first'])}-{int(cr['year_last'])}",
        n_DSK=int(cr["n"]), n_ortaq=len(_com),
        nisbət_CV_pct=(None if np.isnan(_cv) else round(_cv, 2)),
        lövbər_nisbəti=(None if np.isnan(_anch) else round(_anch, 4)),
        Q1_vahid=("keçir" if _g1 else "KEÇMİR"),
        Q2a_mütənasiblik=("keçir" if _g2a else "keçmir"),
        Q2b_lövbər_eyniliyi=("keçir" if _g2b else "keçmir"),
        qərar=("DİRÇƏLDİLİR" if _ok else "SABİT QALIR")))

revive_tbl = pd.DataFrame(revive_rows)
print(f"Dirçəldilən məhsul sayı: {len(REBASED)} / {len(CON)}")
print(revive_tbl.to_string(index=False))
print()
print("Uyğunluq statusu (konkordans cədvəlinin öz qeydi):")
for _, cr in CON.iterrows():
    print(f"  sətir {int(cr['vereq_row']):3d}  {cr['vereq_name_az'][:30]:30s} → "
          f"{cr['series_code']:18s} {cr['status_az'][:78]}")
```

**Diaqnostikanın oxunuşu.** Beş məhsulda (şəkər, kərə yağı, bitki yağları, un, xörək duzu) nisbətin variasiyası **sıfıra yaxındır** — yəni şablonun həmin sətirləri elə DSK-nın 18-ci cədvəlinin özüdür, sadəcə istinadı qırılıb. Pendir və kəsmikdə nisbət 1,0005 (CV 1,63 %), mal ətində 0,9737 (CV 0,08 %), quş ətində 0,9032 (CV 0,24 %), emal edilmiş duru süddə 0,4571 (CV 1,85 %) — hamısı sabit əmsallardır və yenidən bazalaşdırma onları düzgün ötürür.

**Çörək (sətir 57) QAPI-2b ilə keçir, 2a ilə yox — və bunun səbəbi şablonun özündədir.** Şablonun çörək sətrində 2017-ci ildə tərif sınığı var (2016: 762,0 → 2017: 1 223,0), ona görə ortaq beş ildə nisbət 0,63-dən 1,0002-yə tullanır və CV 23,6 % olur. Lakin **lövbər ilində iki sıra praktik olaraq eyni ədəddir** (1 223,0 ↔ 1 222,8, fərq 0,02 %), yəni şablonun CARİ tərifi məhz DSK-nın 18-ci cədvəlinin sətridir; qırılmış olan şablonun öz 2013–2016 fraqmentidir. Lövbərin 2017 seçilməsi bu sınığı avtomatik kənarlaşdırır.

**Meyvə-tərəvəz şirələri (sətir 50) DİRÇƏLDİLMİR.** Şablon min dkl, DSK isə min ton ölçür (konkordansın `factor` sahəsi məhz buna görə boşdur — QAPI-1 keçmir) və DSK sətri şirələri konservlərlə birlikdə sayır; ortaq beş ildə nisbət 2,41-dən 24,57-yə qədər dəyişir. Sətir SABİT (2017 səviyyəsi) qalır və Məhdudiyyətlər bölməsində DATA-GAP kimi saxlanılır — burada tələb olunan konkordans deyil, **ölçü vahidi uzlaşdırılmasıdır**.

## 3. Model seçimi

**Sürücü təyinatı.** Hər məhsulun natural həcm artımı (Δln Q) paketin özəyində artıq qiymətləndirilmiş dörd yol ilə izah edilir — məhsul öz FR-inin yenidən qurulmasına ehtiyac qalmadan:

| Sürücü açarı | Mənbə | İqtisadi məna |
|---|---|---|
| `oil_realg` | FR05b (modelləşdirilmiş — əmtəəlik neft və əmtəəlik qazın əvvəlki il qiymətləri ilə çəkilənmiş Laspeyres həcm indeksi) | neft-qaz hasilatının real artımı |
| `nonoil_realg` | FR1 nüvəsi (ansambl) | qeyri-neft ÜDM-in real artımı |
| `g_inv_core` | FR1 nüvəsi (struktur tənlik) | ümumi əsas kapitala yığımın real artımı |
| `g_cons_core` | FR1 nüvəsi (struktur tənlik) | son istehlakın real artımı |

Təyinat **kod daxilində, sətir-sətir əl ilə yazılmadan**, məhsulun mənsub olduğu bölmə kodundan (`section_no`, şablonun öz iyerarxiyası — 1.1, 2.1., 3.14. və s.) hesablanır: neft-qaz hasilatı və neft məhsulları → `oil_realg`; tikinti xammalı və tikinti materialları → `g_inv_core`; sənaye/investisiya girişi olan digər emal sənayesi sahələri (kimya, metallurgiya, maşınqayırma, nəqliyyat vasitələri) → `g_inv_core`; istehlak malları (qida, içki, tütün, toxuculuq, geyim, dəri, kağız, əczaçılıq, mebel, zərgərlik) → `g_cons_core`; kənd təsərrüfatı, mədənçıxarmanın digər sahələri, kommunal xidmətlər, nəqliyyat və rabitə → `nonoil_realg` (aqreqat qeyri-neft yolu, sahə-səviyyəli FR2 hələ qurulmadığı üçün ən yaxşı mövcud vəkalətçi kimi işlədilir — bu, aşağıda Məhdudiyyətlər bölməsində açıq qeyd olunan bir **DEVIASİYA**dır: brif FR2-nin sahə artımının sürücü kimi işlədilməsini nəzərdə tuturdu, lakin FR2 bu tikinti dalğasında (Wave D) hələ mövcud deyil və FR4 ondan asılı deyil (D2 DAG-a görə paralel qrupdadır); ona görə defolt FR1 nüvəsinin aqreqat qeyri-neft yolu olur, FR2 gələcəkdə əlavə olunarsa yüksəldilə bilər).

**Qiymətləndirmə qaydası (hər məhsul üçün, kod daxilində hesablanır):**
1. `n_eff ≥ 8` (sıfır keçidləri çıxıldıqdan sonra etibarlı Δln müşahidə sayı) və sürücü ilə kəsişmə ≥ 8 olduqda: `ols_hac(Δln Q, sürücü)` qiymətləndirilir.
2. **GATE:** əmsal müsbətdirsə VƏ p < 0,10-dursa → **STRUKTUR** üsulu namizədliyə keçir; lakin sürücünün fərziyyə yolu ilə birləşdirildikdə orta illik nəzərdə tutulan artım ±25 %-dən kənara çıxarsa (kiçik nümunədə statistik əhəmiyyət təsadüfən keçilə bilər — bax Bölmə 5-in doğruluq əks-təsdiqi), GATE geri çevrilir və ETALON-a keçilir; əks halda STRUKTUR saxlanılır, proqnoz sürücünün öz yolu ilə (2025 üçün faktiki, 2026–2030 üçün fərziyyə) törədilir.
3. GATE keçilmirsə (statistik olaraq və ya doğruluq əks-təsdiqinə görə) → **ETALON** üsulu: 2019–2024 pəncərəsinin median Δln artımı, illik ±12 % ilə tavanlanır (hər tavanlanma hadisəsi aşağıda cədvəldə göstərilir — heç biri gizlədilmir).
4. `n_eff < 8` olduqda (real müşahidə azlığı) → **SABİT** üsulu: son real dəyərdə düz xətt, DATA-GAP bayrağı ilə. Bütün tarixi dəyərləri sıfır olan tək məhsul (Etilen, sətir 113) ayrıca **STRUKTURAL SIFIR** bayrağı alır.
5. Sətir 179 (adı `#REF!`) tamamilə İSTİSNA edilir — heç bir üsul tətbiq oluna bilməz, çünki nə ad, nə bir dənə belə real dəyər mövcuddur.

**Proqnoz üfüqü — FR4-ə məxsus uzadılma.** Şablonun hesabat hissəsi 2024-də bitdiyi üçün bu FR-in proqnoz üfüqü, digər FR-lərdən fərqli olaraq, **2025–2030** altı ilini əhatə edir: 2025 üçün sürücülərin ARTIQ BİLİNƏN faktiki dəyəri (FR1/FR5-in 2025 faktiki nəticələri) istifadə olunur — bu, gələcəyə baxış deyil, sadəcə şablonun öz məhsul sırasının 2025-ə çatmamasının kompensasiyasıdır.

```python
def classify_driver(section_no):
    # Bölmə kodundan sürücü təyin edir. Bütün 37 bölmə kodu aşağıda əhatə olunur —
    # 'TƏSNİF EDİLMƏDİ' kateqoriyasına düşən sətir qalmamalıdır (aşağıda təsdiqlənir).
    if section_no in ("2.1.", "3.10."):
        return "oil_realg", "neft-qaz hasilatı / neft məhsulları (bölmə kodu)"
    if section_no in ("2.3.", "3.14."):
        return "g_inv_core", "tikinti xammalı / tikinti materialları (bölmə kodu)"
    if section_no in ("3.7.", "3.11.", "3.13.", "3.15.", "3.16.", "3.17.", "3.18.", "3.19.",
                       "3.20.", "3.21.", "3.23."):
        return "g_inv_core", "sənaye/investisiya girişi (bölmə kodu)"
    if section_no in ("3.1.", "3.2.", "3.3.", "3.4.", "3.5.", "3.6.", "3.8.", "3.9.", "3.12.",
                       "3.22.", "3.24."):
        return "g_cons_core", "istehlak malı (bölmə kodu)"
    if section_no in ("1.1", "1.2", "2.2."):
        return "nonoil_realg", "kənd təsərrüfatı/mədənçıxarma — sektor defolt (FR2 hələ mövcud deyil)"
    if section_no in ("4.", "5.", "6.1.1.", "6.1.2.", "6.1.3.", "6.2.1.", "6.2.2.", "6.2.3."):
        return "nonoil_realg", "kommunal/nəqliyyat/rabitə — sektor defolt (FR2 hələ mövcud deyil)"
    return "nonoil_realg", "TƏSNİF EDİLMƏDİ — ümumi qeyri-neft defolt"


p4[["driver", "driver_reason"]] = p4["section_no"].apply(lambda s: pd.Series(classify_driver(s)))
print("Sürücü təsnifatı (179 sətir üzrə):")
print(p4["driver"].value_counts())
print()
unclassified = p4[p4["driver_reason"].str.startswith("TƏSNİF EDİLMƏDİ")]
print(f"'TƏSNİF EDİLMƏDİ' defoltuna düşən sətir sayı: {len(unclassified)} (0 olmalıdır)")
assert len(unclassified) == 0, "bütün bölmə kodları açıq təsnif edilməlidir"
```

## 5. Qiymətləndirmə

Aşağıdakı funksiya sıfır-keçidli sıraların loqarifmik artımını hesablayır (bəzi niş sənaye məhsulları — məsələn Etilen — bəzi illərdə tam sıfır istehsala malikdir; sıfırdan keçən illik keçid loqarifmi təyin olunmadığı üçün NaN kimi atılır, sıfır kimi OXUNMUR).

```python
def log_growth(Q):
    # Sıfır dəyərli illər loqarifmə keçmədən NaN kimi işarələnir və atılır — sıfır heç vaxt
    # süni artım kimi şərh olunmur.
    q = Q.replace(0, np.nan)
    return np.log(q).diff().dropna()


CAP_PCT = 12.0          # illik tavan (%) — ETALON üsulunda hər tavanlanma aşağıda cədvəldə göstərilir
PLAUSIBILITY_PCT = 25.0  # GATE-i keçən STRUKTUR tənliyinin belə hörmətə layiq olması üçün üst hədd:
                         # kiçik nümunədə (n=11) statistik əhəmiyyət təsadüfən keçilə bilər, lakin
                         # nəticə 6 illik üfüqdə iqtisadi cəhətdən inandırıcı olmalıdır — orta illik
                         # nəzərdə tutulan artım bu hədddən kənara çıxarsa, tənlik RƏDD edilir və
                         # ETALON üsuluna keçilir (aşağıda hər hal açıq hesabat olunur).


def median_growth_capped(dlq):
    win = dlq.loc[(dlq.index >= 2019) & (dlq.index <= 2024)]
    if win.empty:
        win = dlq
    med_log = float(win.median())
    med_pct = (np.exp(med_log) - 1) * 100
    capped_pct = float(np.clip(med_pct, -CAP_PCT, CAP_PCT))
    capped_log = float(np.log(1 + capped_pct / 100.0))
    return med_pct, capped_pct, capped_log
```

Əsas dövrə: hər məhsul üçün sürücü ilə qiymətləndirmə (GATE), uğursuzluqda ETALON, kifayət qədər müşahidə olmadıqda SABİT üsulu. Bütün 178 istifadə edilə bilən məhsul üzərində icra olunur (sətir 179 yuxarıda artıq istisna edilib).

```python
results = []
excluded = []

for _, r in p4.iterrows():
    sid, row, label, unit = r["series_id"], int(r["row"]), r["name_az"], r["unit"]
    driver = r["driver"]
    Q = data_layer.vereq_actuals("2.4.1.4.", label, row=row)
    # bu buraxılış: 48-58-ci sətirlərin `#REF!`-lə kəsilmiş sırası DSK mənbəyi ilə əvəz olunur
    # (§2.1-in yenidən bazalaşdırma qaydası). Şablonun öz dəyəri yalnız lövbər kimi işlədilir.
    revived = row in REBASED
    if revived:
        Q_tmpl = Q.copy()
        Q = REBASED[row]
    if len(Q) == 0:
        excluded.append(dict(series_id=sid, row=row, name_az=label, unit=unit,
                             reason="şablonda heç bir real dəyər yoxdur (bütün illər #REF!)"))
        continue

    last_year = int(Q.index.max())
    last_val = float(Q.loc[last_year])
    dlq = log_growth(Q)
    n_eff = len(dlq)
    all_zero = bool((Q == 0).all())

    rec = dict(series_id=sid, row=row, name_az=label, unit=unit, driver=driver,
              driver_reason=r["driver_reason"], n_level=len(Q), n_eff=n_eff,
              last_year=last_year, last_val=last_val, all_zero=all_zero,
              n_hesabat=int(r["n_hesabat"]), n_ref=int(r["n_ref"]),
              res=None, cap_hit=False, revived=revived)

    hist_x = HIST[driver].reindex(dlq.index)
    aligned = pd.concat([dlq.rename("y"), hist_x.rename("x")], axis=1).dropna()

    if n_eff < 8 or len(aligned) < 8:
        rec["method"] = "STRUKTURAL SIFIR" if all_zero else "SABİT (DATA-GAP, n<8)"
        rec["gate_pass"] = None
        dlq_fc = pd.Series({y: 0.0 for y in FY_EXT})
    else:
        res = ek.ols_hac(aligned["y"], aligned[["x"]], maxlags=1, name=sid, verbose=False)
        beta = float(res.params["x"]); pval = float(res.pvalues["x"])
        gate = bool((pval < 0.10) and (beta > 0))
        rec["res"] = res
        rec["gate_pass"] = gate
        rec["plausibility_override"] = False
        struktur_dlq_fc = None
        if gate:
            xpath = {2025: NOW25[driver]}
            for y in FY:
                xpath[y] = float(FUT[driver].loc[y])
            struktur_dlq_fc = pd.Series({y: float(res.params["const"] + beta * xpath[y]) for y in FY_EXT})
            implied_avg_pct = (np.exp(float(struktur_dlq_fc.mean())) - 1) * 100
            rec["implied_avg_pct"] = implied_avg_pct
            if abs(implied_avg_pct) > PLAUSIBILITY_PCT:
                # statistik cəhətdən əhəmiyyətli, lakin iqtisadi cəhətdən inandırıcı olmayan (kiçik
                # nümunədə təsadüfi) nəticə — GATE geri çevrilir, ETALON-a keçilir, hadisə hesabat olunur
                gate = False
                rec["gate_pass"] = False
                rec["plausibility_override"] = True
        if gate:
            rec["method"] = "STRUKTUR"
            dlq_fc = struktur_dlq_fc
        else:
            rec["method"] = "ETALON (median artım, tavanlanmış)"
            med_pct, capped_pct, capped_log = median_growth_capped(dlq)
            rec["cap_hit"] = bool(abs(capped_pct - med_pct) > 1e-9)
            rec["med_pct_raw"] = med_pct
            rec["med_pct_capped"] = capped_pct
            dlq_fc = pd.Series({y: capped_log for y in FY_EXT})

    # səviyyə yolunun tərtibi — son real dəyərdən mürəkkəbləşdirilir
    lvl = {}
    cur = last_val
    for y in FY_EXT:
        cur = cur * float(np.exp(dlq_fc.loc[y]))
        lvl[y] = cur
    rec["level_path"] = pd.Series(lvl)
    rec["growth_path_pct"] = pd.Series({y: (np.exp(dlq_fc.loc[y]) - 1) * 100 for y in FY_EXT})

    try:
        mb = data_layer.vereq_baseline("2.4.1.4.", label, row=row)
    except Exception:
        mb = pd.Series(dtype=float)
    rec["ministry"] = mb

    results.append(rec)

print(f"Qiymətləndirilən məhsul sayı: {len(results)}")
print(f"İstisna edilən məhsul sayı:  {len(excluded)}")
for e in excluded:
    print(" -", e["series_id"], "—", e["reason"])
```

```python
summ = pd.DataFrame([{k: v for k, v in r.items() if k not in ("res", "level_path", "growth_path_pct", "ministry")}
                     for r in results])
print(summ.groupby(["driver", "method"]).size().unstack(fill_value=0))
print()
print(f"Ümumi: {len(summ)} məhsul, {summ['method'].eq('STRUKTUR').sum()} STRUKTUR, "
      f"{summ['method'].str.startswith('ETALON').sum()} ETALON, "
      f"{summ['method'].str.startswith('SABİT').sum() + summ['method'].eq('STRUKTURAL SIFIR').sum()} SABİT/sıfır.")
```

**İllüstrativ tənliklər.** Üç sürücü qrupundan bir-bir nümunə: xam neft hasilatı (`oil_realg`), sement istehsalı (bir ETALON nümunəsi olaraq — GATE-i keçmədiyi üçün median üsulla), mal əti istehsalı (`g_cons_core`, STRUKTUR nümunəsi olaraq mövcud olduqda). Əmsallar HAC standart xətaları ilə göstərilir.

```python
from IPython.display import Math, display

def eq_latex(res, driver_label, lhs="\\Delta\\ln Q_t"):
    b, se_b = res.params["x"], res.se["x"]
    c, se_c = res.params["const"], res.se["const"]
    sign = "+" if b >= 0 else "-"
    body = (r"\underset{(" + f"{se_c:.4f}" + r")}{" + f"{c:.4f}" + "}"
           + f" {sign} " + r"\underset{(" + f"{se_b:.4f}" + r")}{" + f"{abs(b):.4f}" + r"}\," + driver_label)
    return lhs + " = " + body

DRV_LABEL = {"oil_realg": r"g^{oil}_t", "nonoil_realg": r"g^{qeyri\text{-}neft}_t",
            "g_inv_core": r"g^{inv}_t", "g_cons_core": r"g^{cons}_t"}

oil_rec = next(r for r in results if r["series_id"] == "v4_r035_minton")
print(f"Xam neft hasilatı — {oil_rec['method']}, n = {oil_rec['res'].n}, "
      f"nümunə {oil_rec['res'].sample[0]}-{oil_rec['res'].sample[1]}, "
      f"düzəldilmiş R² = {oil_rec['res'].adj_r2:.3f}")
display(Math(eq_latex(oil_rec["res"], DRV_LABEL[oil_rec["driver"]])))
```

179 məhsulun hamısını genişlənən pəncərəli sınaqdan keçirmək (hər biri üçün hər il yenidən qiymətləndirmə) hesablama baxımından israfçıdır və məlumat verici deyil — sınaq pəncərəsi hər bir 11-12 illik sıra üçün eyni dərəcədə qısadır. Bunun əvəzinə brifin göstərdiyi kimi **təmsilçi panel** seçilir: xam neft, təbii qaz, avtobenzin, dizel yanacağı (dördü də neft-qaz zənciri), sement (tikinti), buğda (kənd təsərrüfatı), elektrik enerjisi (kommunal), üstəgəl həcminə görə beş əlavə iri göstərici (mobil telefon abunəçiləri, nəqliyyat sektorunda yük daşınması və dövriyyəsi — cəmi, sərnişin dövriyyəsi, yumurta istehsalı) — cəmi 12 məhsul, altı bölmənin hamısından təmsil ilə.

Hər panel üzvü üçün **STRUKTUR** modeli (hər buraxılışda yenidən qiymətləndirilir, brifin D4 qaydasına uyğun) **RW** (təsadüfi gəzişmə) etalonuna qarşı sınaqdan keçirilir.

```python
PANEL_ROWS = [35, 36, 98, 100, 130, 7, 182, 217, 190, 200, 210, 24]
p4_by_row = p4.set_index("row")

panel_bt_tables = []
panel_summary = []

for row in PANEL_ROWS:
    rr = p4_by_row.loc[row]
    sid, label, driver = rr["series_id"], rr["name_az"], rr["driver"]
    Q = data_layer.vereq_actuals("2.4.1.4.", label, row=row)
    dlq = log_growth(Q)
    hist_x = HIST[driver].reindex(dlq.index)
    F = pd.concat([dlq.rename("y"), hist_x.rename("x")], axis=1).dropna()

    def _make_struktur(driver_series):
        def _fn(train, target_year):
            d = train.dropna()
            if len(d) < 6:
                return np.nan
            res_v = ek.ols_hac(d["y"], d[["x"]], maxlags=1, verbose=False)
            if target_year not in driver_series.index:
                return np.nan
            xv = float(driver_series.loc[target_year])
            return float(res_v.params["const"] + res_v.params["x"] * xv)
        return _fn

    bt = ek.backtest(F, {"STRUKTUR": _make_struktur(hist_x)}, min_train=6, min_points=3,
                     series_code=sid, verbose=False)
    panel_bt_tables.append(bt.table)
    # DÜZƏLİŞ. Əvvəllər bu sütuna `bt.skill` yazılırdı — bu, ANSAMBLIN bacarığıdır, STRUKTUR
    # modelininki DEYİL. `ek.backtest` ansambla yalnız RW-ni üstələyən modeli buraxır; STRUKTUR
    # uduzduqda çəki tam RW-yə keçir və ansambl bacarığı MƏNFİ ola bilmir, 0-a bərabər olur.
    # Nəticədə sement kimi hal (STRUKTUR RMSE 0,131 vs RW 0,011) cədvəldə "0,000" kimi görünür,
    # halbuki modelin öz bacarığı −1093 %-dir; panelin "orta bacarığı" da bu döşəmə səbəbindən
    # yuxarı qərəzli idi. İndi hər iki kəmiyyət ayrıca verilir.
    _rw, _sr = bt.rw_rmse, bt.rmse.get("STRUKTUR")
    _sk_struct = (None if (_rw is None or not _rw or _sr is None) else float(1.0 - _sr / _rw))
    panel_summary.append(dict(series_id=sid, name_az=label, driver=driver, n_eff=len(F),
                              n_common=len(bt.common_years), rw_rmse=_rw,
                              struktur_rmse=_sr,
                              struktur_bacarigi=_sk_struct,      # modelin ÖZ bacarığı (mənfi ola bilər)
                              ansambl_bacarigi=bt.skill))        # RW-döşəməli ansambl bacarığı

panel_bt_all = pd.concat(panel_bt_tables, ignore_index=True) if panel_bt_tables else pd.DataFrame()
panel_summ = pd.DataFrame(panel_summary).round(
    {"rw_rmse": 4, "struktur_rmse": 4, "struktur_bacarigi": 3, "ansambl_bacarigi": 3})
print(panel_summ.to_string(index=False))
print()
_vs = panel_summ["struktur_bacarigi"].dropna()
_beat = int((panel_summ["struktur_rmse"] < panel_summ["rw_rmse"]).sum())
print(f"STRUKTUR modelinin ÖZ bacarığı — median {_vs.median()*100:+.1f} %, "
      f"orta {_vs.mean()*100:+.1f} %; {_beat}/{len(panel_summ)} məhsulda RW üstələnir.")
print("Orta ilə median arasındakı fərq bir müşahidədən gəlir: sementdə RW RMSE-si "
      f"{float(panel_summ.loc[panel_summ.series_id=='v4_r130_minton','rw_rmse'].iloc[0]):.4f} "
      "kimi çox kiçikdir (həcm beş il praktik olaraq dəyişmir), ona görə istənilən model "
      "nisbətdə çox pis görünür — bacarığı "
      f"{float(panel_summ.loc[panel_summ.series_id=='v4_r130_minton','struktur_bacarigi'].iloc[0])*100:+.0f} %. "
      "Median bu təhrifə davamlıdır və panelin əsl mənzərəsini verir.")
print("QEYD (şərtilik): STRUKTUR modeli proqnoz anında sürücünün HƏMİN İLKİ faktiki dəyərini "
      "bilir (proqnozda da sürücü fərziyyə yolu ilə verilir, yəni ölçmə istifadə ilə eyni "
      "rejimdədir), RW isə heç nə bilmir — müqayisə bu mənada STRUKTUR-un xeyrinədir.")
```

**Ölçünün özü bu buraxılışda düzəldilib.** Əvvəlki buraxılışda cədvəlin `skill` sütununa
`ek.backtest`-in ANSAMBL bacarığı yazılırdı, halbuki sütun STRUKTUR modelinin öz bacarığı kimi
oxunurdu. Fərq mexanikidir: ansambl yalnız RW-ni üstələyən modeli saxlayır, uduzduqda çəki tam
RW-yə keçir və ansambl bacarığı mənfi ola bilmir — 0-a bərabər olur. Ona görə sement sətri
"0,000" görünürdü, halbuki modelin öz bacarığı **−1093 %**-dir, və panelin elan olunan "orta
bacarığı" (+27,4 %) bu döşəmə səbəbindən yuxarı qərəzli idi.

**Düzəldilmiş oxunuş.** STRUKTUR modelinin öz bacarığı: **median +29,9 %**, **12 məhsuldan
11-ində RW üstələnir**. Orta isə **−63,7 %**-dir və bu rəqəm tək bir müşahidədən gəlir: sementin
həcmi sınaq pəncərəsində praktik olaraq sabit qaldığı üçün RW RMSE-si 0,011 kimi çox kiçikdir və
istənilən model ona nisbətdə fəlakətli görünür. Belə halda median düzgün xülasə ölçüsüdür; hər
iki rəqəm çap olunur, heç biri gizlədilmir.

Məzmun tərəfdə nəticə dəyişmir: neft-qaz zənciri məhsullarında (xam neft +55 %, dizel +35 %,
qaz +29 %, avtobenzin +24 %) sürücü RW-ni aydın üstələyir; sement isə üstələmir — bu, GATE-in
sement üçün ETALON-a keçməsi ilə tam uyğundur (Bölmə 5).

**Şərtilik qeydi.** STRUKTUR modeli proqnoz anında sürücünün həmin ilki FAKTİKİ dəyərini bilir,
RW isə heç nə bilmir. Bu, proqnozda sürücünün fərziyyə yolu ilə verilməsi ilə eyni məlumat
rejimidir (yəni ölçmə istifadə ilə uyğundur), lakin müqayisəni STRUKTUR-un xeyrinə çevirir və
bacarıq rəqəmi belə oxunmalıdır.

```python
outputs.write_backtest(panel_bt_all)
print(f"validation_backtest.csv-ə yazıldı: {len(panel_bt_all)} sətir ({len(PANEL_ROWS)} məhsul, panel)")
```

### 6.1. Dirçəldilən qida məhsullarının ayrıca sınağı

Dirçəliş yalnız o zaman mənalıdır ki, uzun sıra **ölçülə bilən** proqnoz keyfiyyəti versin. Ona görə on dirçəldilən sətir üçün ayrıca genişlənən pəncərəli sınaq aparılır: **STRUKTUR** (sətrin öz sürücüsü ilə, hər buraxılışda yenidən qiymətləndirilir) və **ETALON_MEDIAN** (tavanlanmış median artım qaydası) **RW**-yə qarşı. Əvvəlki buraxılışda bu sınaq **mümkün deyildi** — dörd Δln müşahidəsi ilə heç bir pəncərə açılmırdı.

```python
revived_bt_tables, revived_summary = [], []
for _row in sorted(REBASED):
    _rr = p4_by_row.loc[_row]
    _sid, _lab, _drv = _rr["series_id"], _rr["name_az"], _rr["driver"]
    _dlq = log_growth(REBASED[_row])
    _hx = HIST[_drv].reindex(_dlq.index)
    _F = pd.concat([_dlq.rename("y"), _hx.rename("x")], axis=1).dropna()

    def _mk_median(train, t):
        s = train["y"].dropna()
        if len(s) < 6:
            return np.nan
        win = s.loc[(s.index >= int(t) - 6) & (s.index <= int(t) - 1)]
        if win.empty:
            win = s
        med_pct = (np.exp(float(win.median())) - 1) * 100
        return float(np.log(1 + float(np.clip(med_pct, -CAP_PCT, CAP_PCT)) / 100.0))

    _bt = ek.backtest(_F, {"STRUKTUR": _make_struktur(_hx), "ETALON_MEDIAN": _mk_median},
                      min_train=8, min_points=4, series_code=_sid, verbose=False)
    revived_bt_tables.append(_bt.table)
    revived_summary.append(dict(
        series_id=_sid, məhsul=_lab[:28], sürücü=_drv, n_eff=len(_F),
        pəncərə=(f"{_bt.common_years[0]}-{_bt.common_years[-1]}" if _bt.common_years else "boş"),
        n_ortaq=len(_bt.common_years), rw_rmse=_bt.rw_rmse,
        struktur_rmse=_bt.rmse.get("STRUKTUR"), median_rmse=_bt.rmse.get("ETALON_MEDIAN"),
        ansambl_bacarığı_pct=(None if _bt.skill is None else _bt.skill * 100),
        qalib=(", ".join(_bt.weights) if _bt.weights else "RW")))

revived_bt_all = pd.concat(revived_bt_tables, ignore_index=True)
revived_summ = pd.DataFrame(revived_summary).round(
    {"rw_rmse": 4, "struktur_rmse": 4, "median_rmse": 4, "ansambl_bacarığı_pct": 1})
print(revived_summ.to_string(index=False))
_sk = revived_summ["ansambl_bacarığı_pct"].dropna()
print()
print(f"Dirçəldilən sətirlərin orta ansambl bacarığı: {_sk.mean():.1f} % "
      f"({int((_sk > 0).sum())}/{len(_sk)} məhsulda RW üstələnir)")
outputs.write_backtest(revived_bt_all)
print(f"validation_backtest.csv-ə yazıldı: {len(revived_bt_all)} sətir "
      f"({len(REBASED)} dirçəldilən məhsul)")
```

Nəticə olduğu kimi verilir: bu on məhsulun natural həcmi qısa üfüqdə əsasən inersial hərəkət edir, ona görə bir hissəsində təsadüfi gəzişmə üstün qalır. Vacib olan odur ki, **artıq ölçülə bilir** — əvvəlki buraxılışda bu sətirlərin proqnozu 2017-ci ilin dəyərində donmuş düz xətt idi və heç bir sınaq rəqəmi daşımırdı.
